### 5.1 The Leap to Procedural Abstraction

In Chapters 3 and 4, we examined sequential execution, variable binding,
and conditional branching. However, our programs remained monomorphic
and fixed to specific inputs. To double a number, we had to duplicate
the multiplication logic; to compute the roots of a quadratic equation
across different coefficients, we had to duplicate the discriminant
calculation.

To scale software without duplication, we need **procedural
abstraction**.

A **function** is a self-contained computational recipe that
parameterises an expression or sequence of operations over symbolic
placeholders. It decouples the *definition* of an algorithm from its
*invocation*. Once defined, a function (such as that shown in
Figure 5–1) suppresses operational detail: the caller treats it as a
black box mapping inputs to outputs, trusting its internal logic without
needing to re-evaluate its implementation.

In this chapter, we will deconstruct how functions execute in modern
runtimes. We will examine the mechanics of **lexical scoping**,
parameter binding, and the **name-to-body rule** (the principle of
substituting actual arguments for parameter instances within an

### 5.2 Anatomy of a Function: Parameters, Arguments, and Scope

Formally, a function $f$ defines a mapping from a domain set $X$ to a
codomain set (also known as range) $Y$ (see Figure 5–2):

<span id="eq:e5-1">$$f: X \longrightarrow Y\qquad{(5–1)}$$</span>

In software systems, a function consists of three components:

1.  **Formal Parameters**: The ordered sequence of symbolic identifiers
    $(p_1, p_2, \dots, p_k)$ that act as local variables within the
    function body.
2.  **Function Body**: The expression or statement sequence
    $e_{\text{body}}$ that computes the result using the formal
    parameters.
3.  **Defining Lexical Scope**: The enclosing environment
    $\sigma_{\text{def}}$ that existed at the exact moment the function
    was instantiated.

It is essential to maintain a strict distinction between **parameters**
and **arguments**:

- **Parameter** (or formal parameter): The formal identifier declared in
  the function's signature (e.g., `a, b, c`).
- **Argument** (or actual parameter): The concrete value or expression
  supplied by the caller during an invocation (e.g.,
  `multiplyAndAdd(2, 3, 4)` passes arguments $2, 3, 4$).

### 5.3 Scoping Models: Lexical (Static) versus Dynamic

The most critical semantic rule governing functions is how non-local
variables (identifiers referenced inside a function body that are not
declared as formal parameters) are resolved.

#### Lexical (Static) Scoping

In **lexical scoping** (used by Python, Scheme, TypeScript, C++, Rust,
and virtually all modern general-purpose languages), free variables are
resolved according to the physical nesting of scopes in the source text
at **definition time**.

Where a function is called is completely irrelevant to how it resolves
variables; only where it was *defined* matters.

``` python
a = 10

def printX():
  print(a) # Resolves to 10 via lexical definition context

def callerScope():
  a = 99
  printX() # In Lexical Scoping, outputs 10, NOT 99!

printX()
callerScope()
```

#### Dynamic Scoping

In **dynamic scoping** (used in early Lisp dialects, Emacs Lisp, and
Bash shell scripts), free variables are resolved by searching backward
up the active runtime call stack at **invocation time**.

Under dynamic scoping, `printX()` in the snippet above would output `99`
because `callerScope` was the most recent activation frame on the call
stack containing a definition for `x`.

Dynamic scoping makes modular software engineering difficult because a
function's behavior can be silently altered depending on the internal
variable names of whatever code happens to invoke it.

### 5.4 Closures: The Marriage of Code and Environment

When functions are **first-class citizens**—meaning they can be passed
as arguments, returned from other functions, and assigned to data
structures—lexical scoping requires that functions carry their birth
environment with them.

A **closure** is a runtime record pairing a function's code (parameters
and body) with a reference to the lexical environment in which the
function was instantiated:

<span id="eq:e5-2">$$\operatorname{Closure} = \langle \operatorname{Parameters}, \, \operatorname{Body}, \, \sigma_{\text{captured}} \rangle\qquad{(5–2)}$$</span>

``` python
def makeAdder(x: int): # x is captured from this lexical frame
    return lambda y: x + y
    # x remains alive inside the closure even after makeAdder returns

addFive = makeAdder(5)
addTen = makeAdder(10)

print(addFive(3), addTen(3))
```

When `makeAdder(5)` finishes executing, its activation frame cannot be
discarded from memory if the inner lambda references `x`. The inner
function forms a closure that retains a permanent pointer to the
environment frame containing `x = 5`.

### 5.5 The Name-to-Body Rule and $\beta$-Reduction

How does a runtime execute a function call $f(a_1, a_2, \dots, a_k)$?

In the pure substitutional model, function application is governed by
Alonzo Church's **$\beta$-reduction** ($\beta$-conversion) rule:

<span id="eq:e5-3">$$(\lambda (p_1, \dots, p_k). \, e_{\text{body}}) \, (v_1, \dots, v_k) \longrightarrow e_{\text{body}}[p_1 \mapsto v_1, \, \dots, \, p_k \mapsto v_k]\qquad{(5–3)}$$</span>

#### The Name-to-Body Rule

In an environment-based notional machine, the **Name-to-Body Rule**
proceeds through four discrete semantic steps:

1.  **Evaluate the Callee**: Evaluate the operator expression to obtain
    a `Closure` record
    $\langle (p_1, \dots, p_k), e_{\text{body}}, \sigma_{\text{captured}} \rangle$.
2.  **Evaluate the Arguments**: Evaluate each actual argument expression
    $a_i$ in the current caller environment $\sigma_{\text{caller}}$ to
    produce terminal values $(v_1, \dots, v_k)$.
3.  **Instantiate Activation Frame**: Allocate a fresh environment frame
    $\sigma_{\text{call}}$ whose parent pointer points directly to
    $\sigma_{\text{captured}}$ (not $\sigma_{\text{caller}}$!).
4.  **Bind Parameters**: Populate $\sigma_{\text{call}}$ by binding each
    formal parameter name $p_i$ to its corresponding argument value
    $v_i$:
    <span id="eq:e5-4">$$\sigma_{\text{call}} = \sigma_{\text{captured}} \cup \{ p_1 \mapsto v_1, \, \dots, \, p_k \mapsto v_k \}\qquad{(5–4)}$$</span>
5.  **Evaluate Body**: Execute $e_{\text{body}}$ under the newly
    constructed environment $\sigma_{\text{call}}$ and return the
    resulting value.

### 5.6 The Call Stack and Activation Frames

When a function invokes another function (or calls itself recursively),
the runtime manages execution context using a **Call Stack**. Each
active invocation is represented by an **Activation Frame** (or **Stack
Frame**, see Figure 5–3).

An activation frame stores:

- The actual argument values bound to formal parameters.
- Local variables declared within the function body.
- The **Return Address**: The instruction point in the caller to which
  control must return upon completion.
- A **Static Link**: A pointer to the enclosing lexical environment (for
  closure variable access).
- A **Dynamic Link**: A pointer to the caller's stack frame below it on
  the stack.

When a function call completes, its frame is **popped** from the call
stack, restoring the caller's frame as the active execution context. If
a program invokes functions indefinitely without returning (e.g.,
infinite recursion), the stack exceeds its memory limit, triggering a
**Stack Overflow**.

### 5.8 Pure Functions, Referential Transparency, and Composition

In modern system design, functions are categorised based on their
interaction with system state:

#### Pure Functions

A function $f$ is **pure** if and only if it satisfies two criteria:

1.  **Determinism**: For any identical input tuple $(x_1, \dots, x_k)$,
    $f(x_1, \dots, x_k)$ always returns the exact same output $y$. It
    depends on no hidden or external mutable state (such as system
    clocks, global variables, or network responses).
2.  **Freedom from Side Effects**: Evaluating $f$ causes no observable
    mutation to system state outside its local activation frame (no I/O
    operations, disk writes, or global variable mutations).

Pure functions possess **referential transparency**: any call to $f(x)$
can be replaced with its computed value without changing the behavior of
the program.

#### Functional Composition

When functions are pure, they can be composed algebraically. The
composition of two functions $f: Y \to Z$ and $g: X \to Y$ yields a new
function $(f \circ g): X \to Z$:

<span id="eq:e5-5">$$(f \circ g)(x) = f(g(x))\qquad{(5–5)}$$</span>

In Python, functional composition pipelines data through a series of
transformations:

``` python
def compose(f, g):
    return lambda x: f(g(x))

trim = lambda s: s.strip()
toUpper = lambda s: s.upper()
exclaim = lambda s: str(f'{s}!')

formatShout = compose(exclaim, compose(toUpper, trim))
print(formatShout('   hello world   ')) # Outputs: "HELLO WORLD!"
```

Listing 5–1: AST Definition with Functional Nodes (ast_nodes.py):

````
"""
Abstract Syntax Tree (AST) definitions for functions, lambdas, and invocations.
"""

from __future__ import annotations
from dataclasses import dataclass
from typing import Literal, Sequence, TypeAlias

PrimitiveValue: TypeAlias = int | float | bool | str

BinaryOp: TypeAlias = Literal['+', '-', '*', '/', '==', '<', '&&']

@dataclass(frozen=True, slots=True)
class LiteralExpr:
    value: PrimitiveValue
    kind: Literal['literal'] = 'literal'

@dataclass(frozen=True, slots=True)
class VariableExpr:
    name: str
    kind: Literal['variable'] = 'variable'

@dataclass(frozen=True, slots=True)
class BinaryExpr:
    op: BinaryOp
    left: Expr
    right: Expr
    kind: Literal['binary'] = 'binary'

@dataclass(frozen=True, slots=True)
class LetExpr:
    name: str
    initializer: Expr
    body: Expr
    kind: Literal['let'] = 'let'

@dataclass(frozen=True, slots=True)
class IfExpr:
    condition: Expr
    then_branch: Expr
    else_branch: Expr
    kind: Literal['if'] = 'if'

@dataclass(frozen=True, slots=True)
class LambdaExpr:
    """
    Anonymous function declaration: (p1, p2, ...) => body
    """
    parameters: tuple[str, ...]
    body: Expr
    kind: Literal['lambda'] = 'lambda'

@dataclass(frozen=True, slots=True)
class CallExpr:
    """
    Function invocation: callee(arg1, arg2, ...)
    """
    callee: Expr
    arguments: tuple[Expr, ...]
    kind: Literal['call'] = 'call'

Expr: TypeAlias = (
    LiteralExpr
    | VariableExpr
    | BinaryExpr
    | LetExpr
    | IfExpr
    | LambdaExpr
    | CallExpr
)

class Fn:
    """Smart constructors (factories) for Functional AST nodes."""

    @staticmethod
    def lit(value: PrimitiveValue) -> LiteralExpr:
        return LiteralExpr(value)

    @staticmethod
    def var(name: str) -> VariableExpr:
        return VariableExpr(name)

    @staticmethod
    def bin(op: BinaryOp, left: Expr, right: Expr) -> BinaryExpr:
        return BinaryExpr(op, left, right)

    @staticmethod
    def let(name: str, initializer: Expr, body: Expr) -> LetExpr:
        return LetExpr(name, initializer, body)

    @staticmethod
    def if_(condition: Expr, then_branch: Expr, else_branch: Expr) -> IfExpr:
        return IfExpr(condition, then_branch, else_branch)

    @staticmethod
    def lambda_(parameters: Sequence[str], body: Expr) -> LambdaExpr:
        return LambdaExpr(tuple(parameters), body)

    @staticmethod
    def call(callee: Expr, *args: Expr) -> CallExpr:
        return CallExpr(callee, tuple(args))
````

Listing 5–2: Runtime Closures and the Environment Model:

````
"""
Runtime representation of lexical environments, parent pointer chains,
and first-class closures.
"""

from __future__ import annotations
from typing import TypeAlias
#from ast_nodes import Expr, PrimitiveValue

class Closure:
    """
    Represents a runtime Closure pairing formal parameters and function body
    with its captured lexical environment frame.
    """

    def __init__(
        self,
        parameters: tuple[str, ...],
        body: Expr,
        captured_environment: Environment,
        name_tag: str | None = None,
    ) -> None:
        self.parameters: tuple[str, ...] = parameters
        self.body: Expr = body
        self.captured_environment: Environment = captured_environment
        self.name_tag: str | None = name_tag

    def __str__(self) -> str:
        name = f"[Closure: {self.name_tag}]" if self.name_tag else "[Anonymous Closure]"
        return f"{name}({', '.join(self.parameters)})"

    def __repr__(self) -> str:
        return self.__str__()

RuntimeValue: TypeAlias = PrimitiveValue | Closure

class Environment:
    """
    Scoped lexical environment frame with parent pointer hierarchy.
    """

    def __init__(self, parent: Environment | None = None) -> None:
        self._store: dict[str, RuntimeValue] = {}
        self.parent: Environment | None = parent

    def define(self, name: str, value: RuntimeValue) -> None:
        self._store[name] = value

    def lookup(self, name: str) -> RuntimeValue:
        if name in self._store:
            return self._store[name]
        if self.parent is not None:
            return self.parent.lookup(name)
        raise NameError(f"Runtime Error: Unbound identifier '{name}'.")

    def snapshot(self) -> dict[str, str]:
        """Produces a snapshot mapping of all visible bindings."""
        result: dict[str, str] = self.parent.snapshot() if self.parent is not None else {}
        for k, v in self._store.items():
            result[k] = str(v)
        return result
````

Listing 5–3: Call Stack Tracker and Execution Engine:

````
"""
Evaluator implementing the Name-to-Body Rule and dynamic activation frame tracking.
"""

from __future__ import annotations
from dataclasses import dataclass
#from ast_nodes import Expr, LiteralExpr, VariableExpr, BinaryExpr, LetExpr, IfExpr, LambdaExpr, CallExpr
#from environment import Environment, RuntimeValue, Closure

@dataclass(frozen=True, slots=True)
class StackFrameInfo:
    frame_id: int
    function_name: str
    bound_arguments: dict[str, RuntimeValue]

class CallStack:
    """Manages the call stack of activation frames, detecting stack overflow."""

    def __init__(self, max_depth: int = 1024) -> None:
        self.max_depth: int = max_depth
        self._frames: list[StackFrameInfo] = []
        self._frame_sequence: int = 0

    def push(
        self, function_name: str, bound_arguments: dict[str, RuntimeValue]
    ) -> StackFrameInfo:
        if len(self._frames) >= self.max_depth:
            raise RecursionError(
                f"Call Stack Overflow: Maximum recursion depth of {self.max_depth} exceeded."
            )
        self._frame_sequence += 1
        frame = StackFrameInfo(
            frame_id=self._frame_sequence,
            function_name=function_name,
            bound_arguments=bound_arguments,
        )
        self._frames.append(frame)
        return frame

    def pop(self) -> None:
        self._frames.pop()

    @property
    def depth(self) -> int:
        return len(self._frames)

    def get_snapshot(self) -> tuple[StackFrameInfo, ...]:
        return tuple(self._frames)

class FunctionInterpreter:
    """
    Evaluator engine implementing static scoping, closure evaluation, and call tracing.
    """

    def __init__(self) -> None:
        self.call_stack: CallStack = CallStack()
        self._max_observed_stack_depth: int = 0

    def get_max_stack_depth(self) -> int:
        return self._max_observed_stack_depth

    def evaluate(self, expr: Expr, env: Environment | None = None) -> RuntimeValue:
        if env is None:
            env = Environment()

        if self.call_stack.depth > self._max_observed_stack_depth:
            self._max_observed_stack_depth = self.call_stack.depth

        match expr:
            case LiteralExpr(value):
                return value

            case VariableExpr(name):
                return env.lookup(name)

            case BinaryExpr(op, left, right):
                left_val = self.evaluate(left, env)
                right_val = self.evaluate(right, env)
                return self._eval_binary(op, left_val, right_val)

            case LetExpr(name, initializer, body):
                init_val = self.evaluate(initializer, env)
                # If binding a closure to an identifier, assign a diagnostic name tag
                if isinstance(init_val, Closure) and init_val.name_tag is None:
                    init_val.name_tag = name
                inner_env = Environment(parent=env)
                inner_env.define(name, init_val)
                return self.evaluate(body, inner_env)

            case IfExpr(condition, then_branch, else_branch):
                cond = self.evaluate(condition, env)
                if not isinstance(cond, bool):
                    raise TypeError(
                        f"Conditional guard must be Boolean, received '{type(cond).__name__}'."
                    )
                return (
                    self.evaluate(then_branch, env)
                    if cond
                    else self.evaluate(else_branch, env)
                )

            case LambdaExpr(parameters, body):
                # Closure creation: Captures current lexical environment by reference
                return Closure(parameters, body, env)

            case CallExpr(callee, arguments):
                # Step 1: Evaluate Callee Expression
                callee_val = self.evaluate(callee, env)
                if not isinstance(callee_val, Closure):
                    raise TypeError(
                        f"Runtime Error: Attempted to invoke non-function value '{callee_val}'."
                    )

                closure = callee_val

                # Step 2: Evaluate Actual Arguments in Current Caller Environment
                if len(arguments) != len(closure.parameters):
                    raise ValueError(
                        f"Arity Mismatch: Function expected {len(closure.parameters)} arguments, "
                        f"received {len(arguments)}."
                    )

                evaluated_args = [self.evaluate(arg, env) for arg in arguments]

                # Step 3 and 4: Instantiate Activation Frame linked to Captured Environment (Name-to-Body Rule)
                activation_frame = Environment(parent=closure.captured_environment)
                arg_record: dict[str, RuntimeValue] = {}

                for param_name, arg_val in zip(closure.parameters, evaluated_args):
                    activation_frame.define(param_name, arg_val)
                    arg_record[param_name] = arg_val

                # Push frame onto Call Stack for monitoring
                func_name = closure.name_tag or 'anonymous'
                self.call_stack.push(func_name, arg_record)

                # Step 5: Evaluate Function Body under Activation Frame
                try:
                    return self.evaluate(closure.body, activation_frame)
                finally:
                    self.call_stack.pop()

    def _eval_binary(
        self,
        op: str,
        left: RuntimeValue,
        right: RuntimeValue,
    ) -> RuntimeValue:
        if isinstance(left, bool) and isinstance(right, bool):
            match op:
                case '&&': return left and right
                case '==': return left == right

        elif (
            isinstance(left, (int, float))
            and isinstance(right, (int, float))
            and not isinstance(left, bool)
            and not isinstance(right, bool)
        ):
            match op:
                case '+': return left + right
                case '-': return left - right
                case '*': return left * right
                case '/':
                    if right == 0:
                        raise ZeroDivisionError("Division by zero.")
                    return (
                        left // right
                        if isinstance(left, int) and isinstance(right, int)
                        else left / right
                    )
                case '==': return left == right
                case '<': return left < right

        raise TypeError(
            f"Illegal binary operation on ({type(left).__name__} {op} {type(right).__name__})."
        )
````

Listing 5–4: Higher-Order Functional Utilities:

````
"""
Higher-Order Function (HOF) utilities: composition, piping, currying, and memoisation.
"""

from __future__ import annotations
import json
from typing import Callable, TypeVar

A = TypeVar('A')
B = TypeVar('B')
C = TypeVar('C')

UnaryFn = Callable[[A], B]

def compose(f: Callable[[B], C], g: Callable[[A], B]) -> Callable[[A], C]:
    """
    Pure Algebraic Function Composition: (f ∘ g)(x) = f(g(x))
    """
    return lambda x: f(g(x))

def pipe(f: Callable[[A], B], g: Callable[[B], C]) -> Callable[[A], C]:
    """
    Left-to-right pipeline: pipe(f, g)(x) = g(f(x))
    """
    return lambda x: g(f(x))

def curry(fn: Callable[[A, B], C]) -> Callable[[A], Callable[[B], C]]:
    """
    Currying transformation for binary functions: f(a, b) => f(a)(b)
    """
    return lambda a: lambda b: fn(a, b)

def memoize(fn: Callable[[A], B]) -> Callable[[A], B]:
    """
    Pure Memoisation Wrapper for deterministic unary functions.
    """
    cache: dict[str, B] = {}

    def memoized_wrapper(arg: A) -> B:
        key = json.dumps(arg, default=str)
        if key in cache:
            return cache[key]
        computed = fn(arg)
        cache[key] = computed
        return computed

    return memoized_wrapper
````

### 5.10 Summary

In this chapter, we investigated the mechanics of functional
abstraction:

1.  **Procedural Abstraction**: Functions decouple the definition of a
    parameterized computation from its execution, enabling code reuse.
2.  **Lexical (Static) Scoping**: Identifiers are resolved based on the
    physical nesting of scopes at definition time, ensuring that
    functions remain modular and independent of caller context.
3.  **Closures**: First-class functions are represented at runtime as
    closures—tuples pairing the function's parameters and body with a
    reference to the enclosing lexical environment frame.
4.  **The Name-to-Body Rule**: Invoking a function evaluates actual
    arguments in the caller's environment, creates a fresh activation
    frame linked to the closure's captured environment, binds formal
    parameters to argument values, and evaluates the body.
5.  **Call Stack Architecture**: Recursive and nested invocations push
    activation frames onto a dynamic call stack. Frame allocations must
    be bounded to prevent stack overflow faults.
