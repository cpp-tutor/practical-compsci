# Chapter 6: Recursion

> *"To iterate is human, to recurse divine."*  
> — **L. Peter Deutsch**

### 6.1 The Power of Self-Reference

In Chapter 5, we introduced the function as a procedural abstraction: a
black box that transforms inputs to outputs via formal parameters. In
every example explored so far, functions operated in a top-down
hierarchy: function $A$ called function $B$, which called function $C$,
which returned a value that propagated back up the call stack.

What happens when a function calls *itself*?

To the uninitiated, self-referential definitions appear circular—a
paradox akin to defining a word in a dictionary using the very word
itself. Yet in computer science and mathematics, self-reference under a
well-founded termination condition is not a paradox; it is
**recursion**, one of the most powerful and expressive mechanisms in
algorithmic thinking.

Recursion allows complex problems to be solved by decomposing them into
smaller instances of the exact same problem. It dispenses with mutable
loop counters and shared state, replacing them with inductive reasoning
and immutable data transformations.

In this chapter, we will formalise the structural mechanics of
recursion. We will trace **recursive call trees**, examine the
difference between linear and tree recursion, implement **structural
induction** over inductively defined singly-linked lists, analyse fast
divide-and-conquer algorithms, and construct a programmatic call-tree
tracer in Python capable of generating visual execution graphs.

1.  **The Base Case (Termination Condition)**:  
    One or more direct, non-recursive branches that return a known value
    immediately for the simplest possible inputs (e.g., $n = 0$, an
    empty list, or a leaf node). Without a base case, recursion executes
    indefinitely until the call stack exhausts its physical memory
    allocation, triggering a **Stack Overflow**.

2.  **The Inductive Step (Recursive Case)**:  
    The branch where the function invokes itself, passing one or more
    *modified arguments*. Crucially, the sub-problems must be strictly
    smaller in magnitude than the original input according to some
    well-defined measure.

3.  **The Progress Invariant (Well-Founded Ordering)**:  
    There must exist a mathematical guarantee that every recursive call
    makes measurable progress toward the base case. If the domain is the
    natural numbers $\mathbb{N}$, the argument must strictly decrease
    toward $0$; if the domain is a recursive data structure, the
    structural depth must strictly decrease.

### 6.3 Linear Recursion versus Tree Recursion

The topological structure of recursive calls fundamentally dictates an
algorithm's time and space complexity.

#### 1. Linear Recursion

A function is **linearly recursive** if it makes at most one recursive
call per activation frame. The call stack grows linearly ($O(N)$) and
unwinds in reverse order.

Consider the classic mathematical factorial function
$n! = \prod_{i=1}^{n} i$:

$$n! = \begin{cases} 1 & \text{if } n = 0 \\ n \times (n - 1)! & \text{if } n > 0 \end{cases}$$

``` python
def factorial(n: int) -> int:
    return 1 if n <= 0 else n * factorial(n - 1)
```

Call stack trace for `factorial(4)`:

```
[Push] factorial(4) = 4 * factorial(3)
  [Push] factorial(3) = 3 * factorial(2)
    [Push] factorial(2) = 2 * factorial(1)
      [Push] factorial(1) = 1 * factorial(0)
        [Push] factorial(0) = 1  <-- Base Case Reached
      [Pop] returns 1  ==> (1 * 1 = 1)
    [Pop] returns 1    ==> (2 * 1 = 2)
  [Pop] returns 2      ==> (3 * 2 = 6)
[Pop] returns 6        ==> (4 * 6 = 24) --> Result: 24
```

#### 2. Tree Recursion

A function is **tree recursive** if an activation frame spawns *two or
more* recursive calls. This branches execution into a tree of concurrent
exploration.

Consider the naive Fibonacci definition:

$$\operatorname{fib}(n) = \begin{cases} 0 & \text{if } n = 0 \\ 1 & \text{if } n = 1 \\ \operatorname{fib}(n - 1) + \operatorname{fib}(n - 2) & \text{if } n > 1 \end{cases}$$

``` python
def fibonacci(n: int) -> int:
    n if n <= 1 else fibonacci(n - 1) + fibonacci(n - 2)
```

Because each non-base node splits into two sub-invocations, the total
number of function calls grows exponentially:

$$\text{Time Complexity} = O(2^N)$$

Tree recursion often results in redundant computation. For instance,
calculating $\operatorname{fib}(5)$ evaluates $\operatorname{fib}(3)$
twice, $\operatorname{fib}(2)$ three times, and $\operatorname{fib}(1)$
five times. Later in this chapter, we will explore techniques to
eliminate this redundancy.

### 6.5 Divide and Conquer: Fast Exponentiation

Recursion provides a natural structure for the **Divide and Conquer**
algorithmic paradigm:

1.  **Divide**: Partition the problem into independent, smaller
    sub-problems.
2.  **Conquer**: Recursively solve each sub-problem until the base case
    is reached.
3.  **Combine**: Merge the sub-problem solutions to construct the answer
    to the original problem.

A classic demonstration of divide-and-conquer is **Fast Modular
Exponentiation** (exponentiation by squaring). Computing $b^n$ via naive
linear recursion takes $O(n)$ multiplication steps:

<span id="eq:eq6-1">$$b^n = b \times b^{n-1}\qquad{(6–1)}$$</span>

However, by observing that $b^n = (b^2)^{n/2}$ when $n$ is even, we can
halve the problem size at each step:

<span id="eq:eq6-2">$$b^n = \begin{cases} 1 & \text{if } n = 0 \\ (b^{n/2})^2 & \text{if } n \text{ is even} \\ b \times b^{n-1} & \text{if } n \text{ is odd} \end{cases}\qquad{(6–2)}$$</span>

This reduces the time complexity from linear $O(n)$ to logarithmic
$O(\log n)$:

$$\text{Steps for } b^{1024} = 10 \text{ multiplications instead of } 1024!$$

### 6.6 Structural Recursion over Inductive Lists

In imperative programming, collections are typically stored as
contiguous dynamic arrays traversed using mutable `for` loops. In
functional programming, collections are modelled as **inductively
defined data structures**.

A **Singly-Linked List** $\mathcal{L}\langle T \rangle$ is defined
algebraically as:

<span id="eq:eq6-3">$$\mathcal{L}\langle T \rangle = \operatorname{Nil} \mid \operatorname{Cons}(T, \mathcal{L}\langle T \rangle)\qquad{(6–3)}$$</span>

- **$\operatorname{Nil}$**: The empty list (Base Case).
- **$\operatorname{Cons}(\text{head}, \text{tail})$**: A constructor
  pairing a single data element of type $T$ (the head) with another list
  of type $\mathcal{L}\langle T \rangle$ (the tail).

### 6.7 Tail Recursion and Accumulator Transformation

In standard recursion, each active call must wait for the child call to
return before it can compute its own final result (e.g., waiting to
multiply $n$ by the result of $\operatorname{factorial}(n - 1)$). This
forces the runtime to retain every stack frame in memory, consuming
$O(N)$ space.

A recursive function is **tail-recursive** if the recursive call is the
absolute final expression evaluated, with no remaining computation
deferred after the call returns:

$$f(x) = f(x')$$

#### The Accumulator Pattern

Any linear recursive function can be mechanically converted into
tail-recursive form by introducing an **accumulator parameter** that
carries the running computation forward into the next frame:

``` python
# Standard Linear Recursion (O(N) stack frames required):
def factorialLinear(n: int) -> int:
    if n <= 0:
        return 1 
    else:
        return n * factorialLinear(n - 1)

# Tail-Recursive Form using Accumulator (O(1) stack space possible):
def factorialTail(n: int, acc: int = 1) -> int:
    if n <= 0:
        return acc
    else:
        return factorialTail(n - 1, n * acc); # Pure tail call
```

When a compiler supports **Tail Call Optimisation (TCO)**, it rewrites
`factorialTail` into a jump instruction in machine code, reusing the
same stack frame and executing with $O(1)$ stack space.

Listing 6–1: Inductive Singly-Linked List Architecture (list_ops.py):

````
"""
Inductive Singly-Linked List data structure and pure structural recursive operations.
"""

from __future__ import annotations
from dataclasses import dataclass
from typing import Callable, Generic, Literal, Sequence, TypeAlias, TypeVar

T = TypeVar('T')
A = TypeVar('A')
B = TypeVar('B')

@dataclass(frozen=True, slots=True)
class Nil:
    """The empty list (Base Case)."""
    kind: Literal['nil'] = 'nil'

NIL = Nil()

@dataclass(frozen=True, slots=True)
class Cons(Generic[T]):
    """An inductive cons-cell pairing a head element with a tail list."""
    head: T
    tail: FList[T]
    kind: Literal['cons'] = 'cons'

# Functional Inductive List Type: FList[T] = Nil | Cons[T]
FList: TypeAlias = Nil | Cons[T]

class LinkedList:
    """Smart constructors and pure structural recursion operators."""

    @staticmethod
    def nil() -> Nil:
        return NIL

    @staticmethod
    def cons(head: T, tail: FList[T]) -> FList[T]:
        return Cons(head, tail)

    @staticmethod
    def from_array(arr: Sequence[T]) -> FList[T]:
        """Constructs an immutable functional list from a native sequence."""
        current: FList[T] = NIL
        for item in reversed(arr):
            current = Cons(item, current)
        return current

    @staticmethod
    def to_array(lst: FList[T]) -> list[T]:
        """Converts an inductive list into a native Python list."""
        result: list[T] = []
        curr = lst
        while isinstance(curr, Cons):
            result.append(curr.head)
            curr = curr.tail
        return result

    @staticmethod
    def length(lst: FList[T]) -> int:
        """Structural Recursion: Computes the length of a list."""
        match lst:
            case Nil():
                return 0  # Base Case
            case Cons(_, tail):
                return 1 + LinkedList.length(tail)  # Inductive Step

    @staticmethod
    def append(list_a: FList[T], list_b: FList[T]) -> FList[T]:
        """Structural Recursion: Concatenates two lists (A ++ B)."""
        match list_a:
            case Nil():
                return list_b  # Base Case
            case Cons(head, tail):
                return Cons(head, LinkedList.append(tail, list_b))

    @staticmethod
    def map(lst: FList[A], fn: Callable[[A], B]) -> FList[B]:
        """Structural Recursion: Transforms each element via a mapping function."""
        match lst:
            case Nil():
                return NIL
            case Cons(head, tail):
                return Cons(fn(head), LinkedList.map(tail, fn))

    @staticmethod
    def filter(lst: FList[T], predicate: Callable[[T], bool]) -> FList[T]:
        """Structural Recursion: Filters elements satisfying a predicate."""
        match lst:
            case Nil():
                return NIL
            case Cons(head, tail):
                rest = LinkedList.filter(tail, predicate)
                return Cons(head, rest) if predicate(head) else rest

    @staticmethod
    def reverse(lst: FList[T]) -> FList[T]:
        """Tail-Recursive Reversal using an accumulator."""
        def rev_helper(remaining: FList[T], acc: FList[T]) -> FList[T]:
            match remaining:
                case Nil():
                    return acc
                case Cons(head, tail):
                    return rev_helper(tail, Cons(head, acc))

        return rev_helper(lst, NIL)

    @staticmethod
    def fold_right(lst: FList[A], zero: B, combine: Callable[[A, B], B]) -> B:
        """
        Right-Associative Fold (Catamorphism).
        fold_right([1, 2, 3], z, f) => f(1, f(2, f(3, z)))
        """
        match lst:
            case Nil():
                return zero
            case Cons(head, tail):
                return combine(head, LinkedList.fold_right(tail, zero, combine))

    @staticmethod
    def fold_left(lst: FList[A], acc: B, combine: Callable[[B, A], B]) -> B:
        """
        Left-Associative Fold with Accumulator (Tail-Recursive).
        fold_left([1, 2, 3], z, f) => f(f(f(z, 1), 2), 3)
        """
        match lst:
            case Nil():
                return acc
            case Cons(head, tail):
                return LinkedList.fold_left(tail, combine(acc, head), combine)
````

Listing 6–2: Fast Divide-and-Conquer Exponentiation (math_ops.py):

````
"""
Recursive mathematical operations comparing Linear O(N) vs Fast Divide-and-Conquer O(log N).
"""

from __future__ import annotations
from dataclasses import dataclass

@dataclass(frozen=True, slots=True)
class ExponentiationMetrics:
    result: int
    multiplication_steps: int

def power_linear(base: int, exp: int) -> ExponentiationMetrics:
    """
    Computes base^exp using Naive Linear Recursion: O(N) multiplication steps.
    """
    steps = 0

    def recurse(b: int, e: int) -> int:
        nonlocal steps
        if e == 0:
            return 1
        steps += 1
        return b * recurse(b, e - 1)

    res = recurse(base, exp)
    return ExponentiationMetrics(result=res, multiplication_steps=steps)

def power_fast(base: int, exp: int) -> ExponentiationMetrics:
    """
    Computes base^exp using Fast Divide-and-Conquer (Repeated Squaring): O(log N) steps.
    """
    steps = 0

    def recurse(b: int, e: int) -> int:
        nonlocal steps
        if e == 0:
            return 1
        if e == 1:
            return b

        if e % 2 == 0:
            steps += 1
            half = recurse(b, e // 2)
            return half * half
        else:
            steps += 2
            half = recurse(b, (e - 1) // 2)
            return b * half * half

    res = recurse(base, exp)
    return ExponentiationMetrics(result=res, multiplication_steps=steps)
````

Listing 6–3: Programmatic Call-Tree Tracer and Execution Visualiser
(tracer.py):

````
"""
Recursive Execution Call-Tree Tracer and ASCII Graph Formatter.
"""

from __future__ import annotations
from dataclasses import dataclass, field
import json
import time
from typing import Any, Callable, TypeVar

R = TypeVar('R')

@dataclass(slots=True)
class TreeNode:
    id: int
    function_name: str
    args: tuple[Any, ...]
    children: list[TreeNode] = field(default_factory=list)
    return_value: Any = None
    duration_ms: float = 0.0

@dataclass(frozen=True, slots=True)
class TreeMetrics:
    total_calls: int
    max_depth: int

class CallTreeTracer:
    """Execution Call-Tree Tracer for instrumenting recursive functions."""

    def __init__(self) -> None:
        self._node_sequence: int = 0
        self._stack: list[TreeNode] = []
        self.root: TreeNode | None = None

    def trace(self, fn_name: str, fn: Callable[..., R]) -> Callable[..., R]:
        """Wraps a recursive function to record its invocation graph."""

        def traced_wrapper(*args: Any) -> R:
            self._node_sequence += 1
            node = TreeNode(
                id=self._node_sequence,
                function_name=fn_name,
                args=args,
            )

            if not self._stack:
                self.root = node
            else:
                parent = self._stack[-1]
                parent.children.append(node)

            self._stack.append(node)
            start_time = time.perf_counter()

            try:
                result = fn(*args)
                node.return_value = result
                node.duration_ms = (time.perf_counter() - start_time) * 1000.0
                return result
            finally:
                self._stack.pop()

        return traced_wrapper

    def render_ascii_tree(
        self,
        node: TreeNode | None = None,
        prefix: str = '',
        is_last: bool = True,
    ) -> str:
        """Renders the recorded call tree into a formatted ASCII tree string."""
        if node is None:
            node = self.root

        if node is None:
            return '[Empty Call Tree]'

        branch = '└── ' if is_last else '├── '
        arg_str = ', '.join(json.dumps(a, default=str) for a in node.args)
        ret_str = json.dumps(node.return_value, default=str)
        line = f"{prefix}{branch}{node.function_name}({arg_str}) => {ret_str}\n"

        next_prefix = prefix + ('    ' if is_last else '│   ')
        children_output = ''

        for i, child in enumerate(node.children):
            is_child_last = i == len(node.children) - 1
            children_output += self.render_ascii_tree(
                child, next_prefix, is_child_last
            )

        return line + children_output

    def get_metrics(self, node: TreeNode | None = None) -> TreeMetrics:
        """Calculates structural metrics (total call count and maximum depth) of the tree."""
        if node is None:
            node = self.root

        if node is None:
            return TreeMetrics(total_calls=0, max_depth=0)

        total = 1
        max_child_depth = 0

        for child in node.children:
            child_metrics = self.get_metrics(child)
            total += child_metrics.total_calls
            if child_metrics.max_depth > max_child_depth:
                max_child_depth = child_metrics.max_depth

        return TreeMetrics(total_calls=total, max_depth=1 + max_child_depth)
````

### 6.9 Summary

In this chapter, we investigated the foundations of recursive execution:

1.  **Well-Founded Recursion**: A valid recursive function must
    establish base cases, execute inductive steps that strictly reduce
    the problem size, and maintain a progress invariant ensuring
    termination.
2.  **Linear versus Tree Recursion**: Linear recursion spawns a single
    recursive branch ($O(N)$ depth); tree recursion branches multiple
    times, generating exponential call trees ($O(2^N)$) unless
    optimized.
3.  **Divide and Conquer**: Splitting sub-problems multiplicatively
    (e.g. repeated squaring) reduces linear complexities ($O(N)$) down
    to logarithmic time ($O(\log N)$).
4.  **Structural Recursion on Lists**: Inductive data structures
    ($\mathcal{L}\langle T \rangle = \operatorname{Nil} \mid \operatorname{Cons}(T, \mathcal{L}\langle T \rangle)$)
    are processed naturally through structural induction, mapping the
    algebraic shape of the data directly to function call trees.
5.  **Tail Call Optimization**: Accumulators transform deferred linear
    recursive computations into immediate tail calls, enabling compilers
    to reuse stack frames and run in $O(1)$ space.
