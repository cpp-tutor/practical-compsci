### 8.1 State Meets Behaviour

In Chapters 5, 6, and 7, we operated primarily in the functional domain.
We defined data as passive, immutable algebraic structures (such as
inductive lists, stacks, and queues) and manipulated that data using
external, pure functions.

This separation of data from behaviour works well for algorithmic
transformations. However, as software systems grow in
complexity—simulating physical environments, managing graphical user
interfaces, or modelling network topologies—treating data as separate
from the operations that govern it can become unwieldy.

If every function in a large system has direct access to the raw
internals of a data structure, maintaining global invariants becomes
difficult. A single erroneous assignment can corrupt internal state,
causing subtle bugs across the entire codebase.

To manage this complexity, we turn to **Object-Oriented Programming
(OOP)** (see Figure 8–1).

The core idea of OOP is to package state and behaviour together into
autonomous entities called **objects**. An object manages its own
internal data, enforces its own invariants, and communicates with other
objects through well-defined **messages** (method invocations).

In this chapter, we will deconstruct the four pillars of object-oriented
design: **encapsulation, abstraction, inheritance, and polymorphism**.
We compare **structural typing** in TypeScript with **nominal typing**
in C++, explore the mechanics of **dynamic dispatch** via virtual method
tables (**vtables**), and construct a vector-graphics rendering platform
that compiles polymorphic shape hierarchies into clean SVG diagrams.

#### 1. Encapsulation (Information Hiding)

Encapsulation bundles internal data attributes with the methods that
operate on them, restricting direct external access to an object's
private state.

By designating fields as `private` (or `readonly`) in languages which
support it (such as in C++ shown below), an object establishes an
**encapsulation boundary**. External code cannot mutate fields
arbitrarily; it must interact through public methods that validate
inputs and preserve internal invariants.

``` cpp
// Invariant Violation in Unencapsulated Code:
class UnsafeBankVault {
  public:
    float balance = 0.0; // External code can set balance = -50000!
};

// Robust Encapsulation:
class EncapsulatedBankVault {
  private:
    float currentBalance = 0.0;

  public: // All subsequent entities are public
    void deposit(float amount) {
      if (amount <= 0.0) throw std::runtime_error("Deposit amount must be positive.");
      currentBalance += amount; // 'this' is implicit in C++
    }

    float getBalance(){
      return currentBalance;
    }
};
```

#### 2. Abstraction (Contract Separation)

Abstraction hides complex implementation details behind a clean, minimal
interface. A client interacting with a `VectorCanvas` needs to know
*what* operations are available (e.g. `canvas.render()`), not *how* the
underlying coordinate transformations or rasterisation pipelines are
implemented.

In C++, abstraction is formalized through abstract base class (ABC)
declarations, defining a behavioral contract without committing to a
concrete implementation.

``` cpp
class GraphicElement {
  private:
    std::string id;

  public:
    GraphicElement(std::string_view id): id{ id } {} // Constructor only initialises 'id'

    virtual float computeArea() = 0; // Pure virtual declaration, no definition in ABC

    virtual ~GraphicElement() {} // Base class destructors should also be virtual
};
```

#### 3. Inheritance (Subtyping and Code Reuse)

Inheritance allows a new class (the **subclass** or derived class) to
inherit fields and methods from an existing class (the **superclass** or
base class). The derived class can extend the base implementation with
new capabilities or override existing methods with specialized behavior.

``` cpp
class CircleElement: public GraphicElement {
  private:
    float radius;

  public:
    CircleElement(std::string_view id, float radius) // Constructor initialises 'radius'
      : GraphicElement{ id }, radius{ radius } {}    // (and also indirectly 'id')

    virtual float computeArea() override { // Implementation for CircleElement
      return M_PI * radius * radius;
    }

    virtual ~CircleElement() {}
};
```

#### 4. Polymorphism (Subtype Polymorphism and Dynamic Dispatch)

Polymorphism (from the Greek *poly* "many" and *morph* "form") allows
objects of different underlying types to be treated uniformly through a
shared interface.

When a method is called on a polymorphic interface reference:

``` cpp
void f() {
  std::vector<GraphicElement*> shapes{
    new CircleElement("a", 5.0),
    new RectangleElement("b", 10.0, 20.0),
    new CircleElement("c", 2.5)
  };
  for (auto *shape : shapes) {
    std::cout << shape->computeArea() << '\n'; // Dispatches to the correct subclass method at runtime!
    delete shape; // Dispatches to correct (virtual) destructor
  }
}
```

The runtime dynamically resolves and invokes the appropriate method
implementation corresponding to the concrete instance. This mechanism is
known as **Dynamic Dispatch**.

- **Nominal Typing**: Type equivalence is determined strictly by
  explicit type names. Two classes with identical fields and methods are
  considered distinct, incompatible types unless an explicit inheritance
  relationship is declared.
- **Structural Typing**: Type equivalence is determined by the shape and
  members of the data. If object $A$ has all the properties required by
  interface $T$ with matching types, $A$ is assignable to $T$,
  regardless of how or where $A$ was declared.

Structural typing provides great flexibility in high-level Python
applications. However, in Part II of this book, we will transition to
**Modern C++**, where nominal typing is enforced at compile time to
optimize memory layout and method dispatch.

### 8.4 The Mechanics of Dynamic Dispatch: The Virtual Table (VTable)

How does a computer know which concrete implementation to execute when
`shape.render()` is called on an abstract `Shape` reference?

In compiled object-oriented languages like C++, this is orchestrated
through a **Virtual Method Table (vtable)** (see Figure 8–4):

1.  **The VTable**: For every class containing virtual methods, the
    compiler constructs a static table of function pointers pointing to
    the concrete method implementations for that class.
2.  **The VPointer (`vptr`)**: Every object instance of that class
    contains an implicit hidden pointer (the `vptr`) stored at byte
    offset zero in its memory layout, which points directly to its
    class's vtable.
3.  **Dispatch Resolution**: When `shape->render()` is executed in
    machine code, the processor:
    - Reads the object's `vptr`.
    - Indexes into the vtable at the fixed slot allocated for `render`.
    - Executes an indirect jump (`CALL`) to the resolved function
      address.

### 8.6 Modern Best Practice: Composition over Inheritance

While inheritance provides a direct path for code reuse, deep class
hierarchies often introduce maintenance challenges: 1. **The Fragile
Base Class Problem**: Modifying a method in a base class can
inadvertently break assumptions in derived subclasses several levels
down. 2. **Tight Coupling**: Subclasses become coupled to the internal
implementation details of their ancestors, hindering refactoring. 3.
**Inflexible Taxonomies**: Rigid "is-a" hierarchies struggle to model
multi-dimensional concerns (e.g. a `FlyingSwimmingBird` attempting to
inherit from both `FlyingCreature` and `SwimmingCreature`).

Modern object-oriented design favours **Composition over Inheritance**:
constructing complex systems by combining small, focused components
("has-a") rather than extending deep class trees ("is-a").

### 8.7 Experimental Implementation in Python

The following listings use the material from this chapter to build a
vector-graphics platform in Python, featuring:

2.  **An Abstract Polymorphic Shape Hierarchy (`Circle`, `Rectangle`,
    `Polygon`, `CompoundShape`)**. defines the abstract base class

    Composite pattern.
3.  **A Vector Canvas Rendering Engine** that compiles shapes into

Listing 8–1: Vector Geometry, Coordinates, and Bounding Boxes
(geometry.py):

````
"""
Immutable 2D Cartesian geometry primitives and Axis-Aligned Bounding Boxes (AABB).
"""

from __future__ import annotations
from dataclasses import dataclass
import math

@dataclass(frozen=True, slots=True)
class Point2D:
    """Immutable 2D Cartesian Coordinate."""
    x: float
    y: float

    def translate(self, dx: float, dy: float) -> Point2D:
        return Point2D(self.x + dx, self.y + dy)

    def scale(self, factor: float, origin: Point2D | None = None) -> Point2D:
        if origin is None:
            origin = Point2D(0.0, 0.0)
        return Point2D(
            origin.x + (self.x - origin.x) * factor,
            origin.y + (self.y - origin.y) * factor,
        )

    def distance_to(self, other: Point2D) -> float:
        dx = self.x - other.x
        dy = self.y - other.y
        return math.hypot(dx, dy)

@dataclass(frozen=True, slots=True)
class BoundingBox:
    """Axis-Aligned Bounding Box (AABB) for spatial geometry."""
    min_x: float
    min_y: float
    max_x: float
    max_y: float

    def __post_init__(self) -> None:
        if self.min_x > self.max_x or self.min_y > self.max_y:
            raise ValueError(
                f"Invalid BoundingBox dimensions: [{self.min_x}, {self.min_y}] "
                f"to [{self.max_x}, {self.max_y}]"
            )

    @property
    def width(self) -> float:
        return self.max_x - self.min_x

    @property
    def height(self) -> float:
        return self.max_y - self.min_y

    def merge(self, other: BoundingBox) -> BoundingBox:
        return BoundingBox(
            min_x=min(self.min_x, other.min_x),
            min_y=min(self.min_y, other.min_y),
            max_x=max(self.max_x, other.max_x),
            max_y=max(self.max_y, other.max_y),
        )
````

Listing 8–2: The Polymorphic Shape Framework (shapes.py):

````
"""
Polymorphic vector shape hierarchy demonstrating encapsulation,
subtyping, and the Composite design pattern.
"""

from __future__ import annotations
from abc import ABC, abstractmethod
from dataclasses import dataclass
from functools import reduce
import math
from typing import Sequence
#from geometry import Point2D, BoundingBox

@dataclass(frozen=True, slots=True)
class StyleAttributes:
    fill_colour: str = '#38bdf8'
    stroke_colour: str = '#0f172a'
    stroke_width: float = 1.5
    opacity: float = 1.0

class Shape(ABC):
    """
    Abstract base class establishing the contract for all vector shapes.
    """

    def __init__(self, id_tag: str, style: StyleAttributes | None = None) -> None:
        self.id_tag: str = id_tag
        self._style: StyleAttributes = style if style is not None else StyleAttributes()

    @abstractmethod
    def compute_area(self) -> float:
        """Computes the enclosed area of the shape."""
        ...

    @abstractmethod
    def compute_perimeter(self) -> float:
        """Computes the total boundary perimeter of the shape."""
        ...

    @abstractmethod
    def get_bounding_box(self) -> BoundingBox:
        """Calculates the minimal enclosing Axis-Aligned Bounding Box."""
        ...

    @abstractmethod
    def render_svg(self) -> str:
        """Serialises the shape into an SVG vector XML element string."""
        ...

    @abstractmethod
    def translate(self, dx: float, dy: float) -> Shape:
        """Returns a new translated instance of the shape."""
        ...

    def _format_style_attributes(self) -> str:
        return (
            f'fill="{self._style.fill_colour}" '
            f'stroke="{self._style.stroke_colour}" '
            f'stroke-width="{self._style.stroke_width}" '
            f'opacity="{self._style.opacity}"'
        )

class Circle(Shape):
    """Concrete Shape: Circle."""

    def __init__(
        self,
        id_tag: str,
        centre: Point2D,
        radius: float,
        style: StyleAttributes | None = None,
    ) -> None:
        super().__init__(id_tag, style)
        if radius <= 0.0:
            raise ValueError(f"Circle radius must be strictly positive, received {radius}.")
        self._centre: Point2D = centre
        self._radius: float = radius

    def compute_area(self) -> float:
        return math.pi * self._radius * self._radius

    def compute_perimeter(self) -> float:
        return 2.0 * math.pi * self._radius

    def get_bounding_box(self) -> BoundingBox:
        return BoundingBox(
            min_x=self._centre.x - self._radius,
            min_y=self._centre.y - self._radius,
            max_x=self._centre.x + self._radius,
            max_y=self._centre.y + self._radius,
        )

    def render_svg(self) -> str:
        return (
            f'<circle id="{self.id_tag}" cx="{self._centre.x}" cy="{self._centre.y}" '
            f'r="{self._radius}" {self._format_style_attributes()} />'
        )

    def translate(self, dx: float, dy: float) -> Circle:
        return Circle(
            self.id_tag,
            self._centre.translate(dx, dy),
            self._radius,
            self._style,
        )

class Rectangle(Shape):
    """Concrete Shape: Rectangle."""

    def __init__(
        self,
        id_tag: str,
        origin: Point2D,
        width: float,
        height: float,
        style: StyleAttributes | None = None,
    ) -> None:
        super().__init__(id_tag, style)
        if width <= 0.0 or height <= 0.0:
            raise ValueError(
                f"Rectangle dimensions must be positive: width={width}, height={height}."
            )
        self._origin: Point2D = origin
        self._width: float = width
        self._height: float = height

    def compute_area(self) -> float:
        return self._width * self._height

    def compute_perimeter(self) -> float:
        return 2.0 * (self._width + self._height)

    def get_bounding_box(self) -> BoundingBox:
        return BoundingBox(
            min_x=self._origin.x,
            min_y=self._origin.y,
            max_x=self._origin.x + self._width,
            max_y=self._origin.y + self._height,
        )

    def render_svg(self) -> str:
        return (
            f'<rect id="{self.id_tag}" x="{self._origin.x}" y="{self._origin.y}" '
            f'width="{self._width}" height="{self._height}" rx="4" '
            f'{self._format_style_attributes()} />'
        )

    def translate(self, dx: float, dy: float) -> Rectangle:
        return Rectangle(
            self.id_tag,
            self._origin.translate(dx, dy),
            self._width,
            self._height,
            self._style,
        )

class CompoundShape(Shape):
    """
    Composite Pattern: CompoundShape treats a group of shapes polymorphically as a single shape.
    """

    def __init__(
        self,
        id_tag: str,
        children: Sequence[Shape],
        style: StyleAttributes | None = None,
    ) -> None:
        super().__init__(id_tag, style)
        if len(children) == 0:
            raise ValueError("CompoundShape must contain at least one child shape.")
        self._children: tuple[Shape, ...] = tuple(children)

    def compute_area(self) -> float:
        return sum(child.compute_area() for child in self._children)

    def compute_perimeter(self) -> float:
        return sum(child.compute_perimeter() for child in self._children)

    def get_bounding_box(self) -> BoundingBox:
        boxes = (child.get_bounding_box() for child in self._children)
        return reduce(lambda acc, b: acc.merge(b), boxes)

    def render_svg(self) -> str:
        rendered_children = "\n    ".join(child.render_svg() for child in self._children)
        return f'<g id="{self.id_tag}">\n    {rendered_children}\n  </g>'

    def translate(self, dx: float, dy: float) -> CompoundShape:
        translated_children = tuple(child.translate(dx, dy) for child in self._children)
        return CompoundShape(self.id_tag, translated_children, self._style)
````

Listing 8–3: The Vector Canvas Renderer (canvas.py):

````
"""
Vector canvas that aggregates heterogeneous shapes and compiles scenes into SVG documents.
"""

from __future__ import annotations
from functools import reduce
#from shapes import Shape
#from geometry import BoundingBox

class VectorCanvas:
    """Canvas managing a scene of polymorphically dispatched shapes."""

    def __init__(self, width: int, height: int, background_colour: str = '#0f172a') -> None:
        self.width: int = width
        self.height: int = height
        self.background_colour: str = background_colour
        self._shapes: list[Shape] = []

    def add_shape(self, shape: Shape) -> None:
        self._shapes.append(shape)

    def compute_total_area(self) -> float:
        return sum(s.compute_area() for s in self._shapes)

    def get_global_bounding_box(self) -> BoundingBox | None:
        if not self._shapes:
            return None
        boxes = (s.get_bounding_box() for s in self._shapes)
        return reduce(lambda acc, b: acc.merge(b), boxes)

    def render_to_svg_document(self) -> str:
        """
        Serialises all polymorphically registered shapes into a standalone SVG document string.
        """
        shape_elements = "\n".join(f"  {shape.render_svg()}" for shape in self._shapes)

        return (
            f'<svg viewBox="0 0 {self.width} {self.height}" width="100%" '
            f'height="{self.height}" xmlns="http://www.w3.org/2000/svg" '
            f'style="background-color: {self.background_colour}; border-radius: 8px;">\n'
            f'  <!-- Rendered Shapes via Polymorphic Dynamic Dispatch -->\n'
            f'{shape_elements}\n'
            f'</svg>'
        )
````

Listing 8–4: Manual Virtual Method Table (VTable) Simulation
(vtable_simulation.py):

````
"""
Explicit simulation of Virtual Method Tables (vtables) and vptr dynamic dispatch.
"""

from __future__ import annotations
from dataclasses import dataclass
import math
from typing import Any, Callable, TypeAlias

AreaFn: TypeAlias = Callable[['RawObjectInstance'], float]
RenderFn: TypeAlias = Callable[['RawObjectInstance'], str]

@dataclass(frozen=True, slots=True)
class VTable:
    """The Virtual Method Table (VTable): A static table of function pointers."""
    compute_area: AreaFn
    render_svg: RenderFn

@dataclass(slots=True)
class RawObjectInstance:
    """
    Raw Object Instance: Models a low-level C++ memory block with a hidden vptr at offset 0.
    """
    vptr: VTable
    fields: dict[str, Any]

# Static VTable Definition for Circle
def _circle_compute_area(self: RawObjectInstance) -> float:
    r: float = self.fields['radius']
    return math.pi * r * r

def _circle_render_svg(self: RawObjectInstance) -> str:
    x = self.fields['x']
    y = self.fields['y']
    r = self.fields['radius']
    return f'<circle cx="{x}" cy="{y}" r="{r}" fill="#38bdf8" />'

CircleVTable = VTable(
    compute_area=_circle_compute_area,
    render_svg=_circle_render_svg,
)

# Static VTable Definition for Rectangle
def _rectangle_compute_area(self: RawObjectInstance) -> float:
    w: float = self.fields['w']
    h: float = self.fields['h']
    return w * h

def _rectangle_render_svg(self: RawObjectInstance) -> str:
    x = self.fields['x']
    y = self.fields['y']
    w = self.fields['w']
    h = self.fields['h']
    return f'<rect x="{x}" y="{y}" width="{w}" height="{h}" fill="#4ade80" />'

RectangleVTable = VTable(
    compute_area=_rectangle_compute_area,
    render_svg=_rectangle_render_svg,
)

# Factory functions allocating raw instances with attached vptrs
def create_raw_circle(x: float, y: float, radius: float) -> RawObjectInstance:
    return RawObjectInstance(
        vptr=CircleVTable,
        fields={'x': x, 'y': y, 'radius': radius},
    )

def create_raw_rectangle(x: float, y: float, w: float, h: float) -> RawObjectInstance:
    return RawObjectInstance(
        vptr=RectangleVTable,
        fields={'x': x, 'y': y, 'w': w, 'h': h},
    )

# Simulated Dynamic Dispatch: Emulates indirect jump through vptr table slots
def dispatch_compute_area(obj: RawObjectInstance) -> float:
    return obj.vptr.compute_area(obj)  # Dereference vptr -> index compute_area -> execute

def dispatch_render_svg(obj: RawObjectInstance) -> str:
    return obj.vptr.render_svg(obj)  # Dereference vptr -> index render_svg -> execute
````

### 8.8 Summary

In this chapter, we explored the foundations of object-oriented design
and dynamic dispatch:

1.  **Encapsulation**: Encapsulating private state behind a public
    method interface protects internal invariants against unauthorized
    mutation.
2.  **Abstraction**: Interfaces establish behavioural contracts that
    decouple what operations an object supports from how those
    operations are implemented.
3.  **Subtype Polymorphism**: Dynamic dispatch allows heterogeneous
    collections of derived objects to be treated uniformly through base
    interfaces.
4.  **Structural versus Nominal Typing**: TypeScript determines type
    compatibility by shape matching (structural), whereas compiled
    languages like C++ enforce identity matching by type name (nominal).
5.  **The Virtual Table (VTable)**: In compiled runtimes, dynamic
    polymorphism is executed via tables of function pointers (vtables)
    indexed through an object's hidden `vptr` at runtime.
6.  **Composition over Inheritance**: Combining focused, autonomous
    components produces more resilient, decoupled architectures than
    deep inheritance hierarchies.
