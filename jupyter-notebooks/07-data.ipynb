### 7.1 The Substance of Computation

In Chapters 1 through 6, we focused on the mechanics of computation: how
bits represent numbers, how logic gates make decisions, how expressions
reduce, how control bifurcates, and how functions execute through
activation frames and self-referential recursion.

Yet pure computational logic without structured information is an engine
running on empty. Real-world software exists to collect, organise,
transform, and retrieve **data**.

How data is structured in memory fundamentally dictates how fast an
algorithm can operate (as outlined in fig. 1). An elegant algorithm
applied to an inappropriate data structure will stall under heavy
workloads; conversely, choosing the right data structure can transform
an unworkable $O(N^2)$ bottleneck into an instantaneous $O(1)$
operation.

In this chapter, we establish the bridge between pure functional
computation and data organisation. We will formalise the concept of
**Abstract Data Types (ADTs)**, dissect the mathematical foundations of
**Big-O asymptotic notation** and **amortised analysis**, examine the
invariants of linear data structures (**Arrays, Stacks, and Queues**),
implement purely functional **persistent data structures** (including
Chris Okasaki’s two-list queue) in Python, and construct a
high-resolution performance benchmarking suite that renders runtime
curves directly into vector graphics.

### 7.2 Abstract Data Types versus Concrete Data Structures

To engineer modular software, we must distinguish between an **Abstract
Data Type (ADT)** and a **Concrete Data Structure**, as shown in fig. 2.

- **Abstract Data Type (ADT)**: A mathematical specification that
  defines a collection of data strictly by its observable *behaviour*,
  supported operations, and algebraic invariants. An ADT specifies
  **what** operations are possible, not **how** they are physically laid
  out in memory.
- **Concrete Data Structure**: The physical memory arrangement and
  algorithmic implementation that satisfies the ADT’s contract.

### 7.3 Algorithmic Complexity and Big-O Asymptotics

When evaluating a data structure, empirical runtimes (measured in
milliseconds) can be misleading. Wall-clock time varies across CPU
architectures, compiler optimisations, operating system scheduling, and
runtime garbage collection pauses.

Computer scientists measure efficiency using **Asymptotic Analysis**,
which characterizes how an algorithm’s resource consumption (time or
memory) scales as the input size $N$ approaches infinity
($N \to \infty$).

#### 1. Formal Definition of Big-O Notation

Let $T(n)$ represent the worst-case execution time of an algorithm on an
input of size $n$. We say that:

$$T(n) \in O(f(n))$$

if and only if there exist positive constants $c > 0$ and $n_0 \ge 0$
such that:

$$T(n) \le c \cdot f(n) \quad \text{for all } n \ge n_0$$

Big-O provides an **asymptotic upper bound**: it guarantees that the
growth rate of the algorithm will not exceed $f(n)$ once the input size
surpasses the threshold $n_0$.

#### 2. The Asymptotic Hierarchy

| Complexity Class | Formal Name | Intuition / Characteristic Example | Growth Scaling ($N = 10^6$) |
|:-----------------|:-----------------|:-----------------|:-----------------|
| $O(1)$ | **Constant** | Hash table lookup, Stack push/pop | $1 \text{ operation}$ |
| $O(\log N)$ | **Logarithmic** | Binary search, balanced tree traversal | $\approx 20 \text{ operations}$ |
| $O(N)$ | **Linear** | Single pass over an array, linear search | $10^6 \text{ operations}$ |
| $O(N \log N)$ | **Linearithmic** | Merge sort, quicksort (average), heap sort | $\approx 2 \times 10^7 \text{ operations}$ |
| $O(N^2)$ | **Quadratic** | Nested loops, bubble sort, pairwise comparisons | $10^{12} \text{ operations (Stalls)}$ |
| $O(2^N)$ | **Exponential** | Naive recursive Fibonacci, subset generation | $\approx 10^{300000} \text{ (Infeasible)}$ |

### 7.4 Amortised Analysis and the Resizing Array

Some operations vary in execution time depending on internal state. For
example, inserting an item into a dynamic array is usually $O(1)$, but
occasionally requires allocating a new memory block and copying all
elements over ($O(N)$).

Evaluating such operations purely by their worst-case cost ($O(N)$) is
overly pessimistic. Instead, we use **Amortised Analysis** to calculate
the average cost per operation over a worst-case sequence of $k$
consecutive operations:

<span id="eq:eq7-1">$$T_{\text{amortised}} = \frac{\sum_{i=1}^{k} \operatorname{Cost}(\text{op}_i)}{k}\qquad{(1)}$$</span>

- Number of regular writes = $N$
- Number of element copies during all resize events:

$$\sum_{j=0}^{k-1} 2^j = 2^k - 1 = N - 1$$

$$\text{Total Work} = N + (N - 1) = 2N - 1 \in O(N)$$

$$\text{Amortised Cost per Insertion} = \frac{2N - 1}{N} < 2 \in O(1)$$

By doubling the capacity geometrically, the expensive $O(N)$ resizing
operations happen infrequently enough that the **amortised cost per
insertion remains strictly $O(1)$**.

#### 1. The Stack (LIFO: Last-In, First-Out)

A **Stack** (fig. 5) enforces the invariant that the last element added
is the first element removed:

- **`push(item)`**: Adds an element to the top of the stack ($O(1)$).
- **`pop()`**: Removes and returns the element at the top of the stack
  ($O(1)$).
- **`peek()`**: Reads the top element without removing it ($O(1)$).

Stacks are used in:

- Runtime activation records and function call stacks.
- Syntax parsing and balancing parentheses/delimiters.
- Depth-First Search (DFS) graph traversals.
- Undo/Redo operational buffers.

#### 2. The Queue (FIFO: First-In, First-Out)

A **Queue** enforces the invariant that the first element added is the
first element removed:

- **`enqueue(item)`**: Appends an element to the rear of the queue
  ($O(1)$).
- **`dequeue()`**: Removes and returns the element at the front of the
  queue ($O(1)$).
- **`peek()`**: Reads the front element without removing it ($O(1)$).

Queues are used in:

- Task schedulers and asynchronous message brokers (e.g. event loops).
- Breadth-First Search (BFS) graph traversals.
- Hardware I/O packet buffers.

### 7.6 Pure Functional Queues: Chris Okasaki’s Two-List Queue

In an imperative language, a queue is typically implemented using a
mutable circular ring buffer in a contiguous array.

However, in purely functional programming, data structures must be
**persistent** (immutable): updating a queue must produce a new queue
while preserving the old version for any code holding an earlier
reference.

#### The Naive Array Trap

Implementing a persistent queue by wrapping an immutable array leads to
performance issues:

- `enqueue(x)`: `[...array, x]` $\implies O(N)$ time (copies the entire
  array).
- `dequeue()`: `array.slice(1)` $\implies O(N)$ time (copies the tail).

#### Okasaki’s Two-List Architecture

In his landmark work *Purely Functional Data Structures* (1996), Chris
Okasaki demonstrated that an immutable queue can achieve **amortised
$O(1)$** performance by representing the queue as a pair of two
immutable singly-linked lists:

$$\mathcal{Q}\langle T \rangle = \langle \text{FrontList}, \, \text{RearList} \rangle$$

1.  **The Front List ($F$)**: Holds elements ready to be dequeued in
    normal FIFO order.
2.  **The Rear List ($R$)**: Holds elements recently enqueued in
    *reverse* order.

#### The Queue Invariant

The data structure maintains a simple invariant: **$F$ is only empty if
$R$ is also empty**.

If a dequeue operation empties $F$ while $R$ still contains elements,
the queue **rotates**: it sets $F' = \operatorname{reverse}(R)$ and
resets $R' = \operatorname{Nil}$.

Although reversing $R$ takes $O(K)$ time, each element is reversed at
most *once* between the time it is enqueued and the time it is dequeued.
Therefore, the **amortised cost per dequeue operation is strictly
$O(1)$**.

Listing 1: Inductive Singly-Linked List Architecture (list_ops.py)

````
"""
Inductive Singly-Linked List data structure backing persistent Stacks and Queues.
"""

from __future__ import annotations
from dataclasses import dataclass
from typing import Callable, Generic, Literal, Sequence, TypeAlias, TypeVar

T = TypeVar('T')

@dataclass(frozen=True, slots=True)
class Nil:
    """The empty list (Base Case)."""
    kind: Literal['nil'] = 'nil'

NIL = Nil()

@dataclass(frozen=True, slots=True)
class Cons(Generic[T]):
    """An inductive cons-cell pairing a head element with a tail list."""
    head: T
    tail: FList[T]
    kind: Literal['cons'] = 'cons'

FList: TypeAlias = Nil | Cons[T]

class LinkedList:
    """Smart constructors and pure structural recursion operators."""

    @staticmethod
    def nil() -> Nil:
        return NIL

    @staticmethod
    def cons(head: T, tail: FList[T]) -> FList[T]:
        return Cons(head, tail)

    @staticmethod
    def from_array(arr: Sequence[T]) -> FList[T]:
        current: FList[T] = NIL
        for item in reversed(arr):
            current = Cons(item, current)
        return current

    @staticmethod
    def to_array(lst: FList[T]) -> list[T]:
        result: list[T] = []
        curr = lst
        while isinstance(curr, Cons):
            result.append(curr.head)
            curr = curr.tail
        return result

    @staticmethod
    def reverse(lst: FList[T]) -> FList[T]:
        def rev_helper(remaining: FList[T], acc: FList[T]) -> FList[T]:
            match remaining:
                case Nil():
                    return acc
                case Cons(head, tail):
                    return rev_helper(tail, Cons(head, acc))

        return rev_helper(lst, NIL)
````

Listing 2: ADT Contracts and Persistent Stack (stack.py)

````
"""
Persistent, purely functional Stack ADT backed by an inductive cons-list.
"""

from __future__ import annotations
from typing import Generic, Protocol, TypeVar
#from list_ops import FList, Cons, Nil, LinkedList, NIL

T = TypeVar('T')

class PersistentStack(Generic[T]):
    """
    Purely functional persistent Stack backed by an inductive singly-linked list.
    All operations provide strict O(1) time and space guarantees via structural sharing.
    """

    def __init__(self, elements: FList[T], count: int) -> None:
        self._elements: FList[T] = elements
        self._count: int = count

    @classmethod
    def empty(cls) -> PersistentStack[T]:
        return cls(NIL, 0)

    def push(self, value: T) -> PersistentStack[T]:
        """O(1) time: Allocate a single cons-cell pointing to the existing list."""
        return PersistentStack(
            LinkedList.cons(value, self._elements),
            self._count + 1,
        )

    def pop(self) -> tuple[T, PersistentStack[T]]:
        """O(1) time: Return head value and tail list reference."""
        match self._elements:
            case Nil():
                raise IndexError("NoSuchElement: Cannot pop from an empty Stack.")
            case Cons(head, tail):
                remaining_stack = PersistentStack(tail, self._count - 1)
                return head, remaining_stack

    def peek(self) -> T:
        match self._elements:
            case Nil():
                raise IndexError("NoSuchElement: Cannot peek into an empty Stack.")
            case Cons(head, _):
                return head

    def is_empty(self) -> bool:
        return self._count == 0

    def size(self) -> int:
        return self._count

    def to_array(self) -> list[T]:
        return LinkedList.to_array(self._elements)
````

Listing 3: Persistent Okasaki Two-List Queue (queue.py)

````
"""
Purely functional persistent Queue implementing Chris Okasaki's two-list algorithm.
"""

from __future__ import annotations
from typing import Generic, Protocol, TypeVar
#from list_ops import FList, Cons, Nil, LinkedList, NIL

T = TypeVar('T')

class PersistentQueue(Generic[T]):
    """
    Purely functional persistent Queue implementing Chris Okasaki's 2-list algorithm.
    Provides amortised O(1) enqueue and dequeue operations.
    """

    def __init__(self, front: FList[T], rear: FList[T], count: int) -> None:
        self._front: FList[T] = front
        self._rear: FList[T] = rear
        self._count: int = count

    @classmethod
    def empty(cls) -> PersistentQueue[T]:
        return cls(NIL, NIL, 0)

    def enqueue(self, value: T) -> PersistentQueue[T]:
        """
        Enqueues an element onto the rear list.
        O(1) worst-case time complexity.
        """
        return self._make_queue(
            self._front,
            LinkedList.cons(value, self._rear),
            self._count + 1,
        )

    def dequeue(self) -> tuple[T, PersistentQueue[T]]:
        """
        Dequeues an element from the front list.
        O(1) amortised time complexity.
        """
        match self._front:
            case Nil():
                raise IndexError("NoSuchElement: Cannot dequeue from an empty Queue.")
            case Cons(head, tail):
                next_queue = self._make_queue(
                    tail,
                    self._rear,
                    self._count - 1,
                )
                return head, next_queue

    def peek(self) -> T:
        match self._front:
            case Nil():
                raise IndexError("NoSuchElement: Cannot peek into an empty Queue.")
            case Cons(head, _):
                return head

    def is_empty(self) -> bool:
        return self._count == 0

    def size(self) -> int:
        return self._count

    def to_array(self) -> list[T]:
        front_array = LinkedList.to_array(self._front)
        rear_array = LinkedList.to_array(LinkedList.reverse(self._rear))
        return front_array + rear_array

    @classmethod
    def _make_queue(
        cls,
        front: FList[T],
        rear: FList[T],
        count: int,
    ) -> PersistentQueue[T]:
        """
        Enforces the Okasaki invariant: Front is only Nil if Rear is also Nil.
        If Front is Nil and Rear is non-empty, rotate:
            Front' = reverse(Rear), Rear' = Nil
        """
        if isinstance(front, Nil):
            return cls(LinkedList.reverse(rear), NIL, count)
        return cls(front, rear, count)
````

Listing 4: Performance Profiling Engine and SVG Chart Generator
(benchmark.py)

````
"""
High-resolution performance benchmarking harness and dynamic SVG line-chart generator.
"""

from __future__ import annotations
from dataclasses import dataclass
import time
from typing import Callable, Sequence

@dataclass(frozen=True, slots=True)
class DataPoint:
    n: int
    duration_ms: float

@dataclass(frozen=True, slots=True)
class BenchmarkSeries:
    name: str
    color: str
    points: tuple[DataPoint, ...]

class PerformanceProfiler:
    """Micro-benchmarking runner using high-resolution performance timers."""

    @staticmethod
    def measure(task: Callable[[], None], warmup_runs: int = 3) -> float:
        """Runs warmup iterations before recording high-resolution elapsed time."""
        for _ in range(warmup_runs):
            task()

        start = time.perf_counter()
        task()
        return (time.perf_counter() - start) * 1000.0  # Duration in milliseconds

    @staticmethod
    def render_svg_chart(
        series_list: Sequence[BenchmarkSeries],
        width: int = 800,
        height: int = 420,
    ) -> str:
        """Generates a complete standalone SVG line chart comparing benchmark series."""
        padding = 60
        chart_width = width - padding * 2
        chart_height = height - padding * 2

        # Determine scale bounds
        max_n = 0
        max_time = 0.0

        for series in series_list:
            for p in series.points:
                if p.n > max_n:
                    max_n = p.n
                if p.duration_ms > max_time:
                    max_time = p.duration_ms

        if max_time == 0.0:
            max_time = 1.0

        def scale_x(n: int) -> float:
            return padding + (n / max_n) * chart_width

        def scale_y(t: float) -> float:
            return height - padding - (t / max_time) * chart_height

        paths_svg: list[str] = []
        legends_svg: list[str] = []

        for idx, series in enumerate(series_list):
            path_coords = [
                f"{'M' if i == 0 else 'L'} {scale_x(p.n):.1f} {scale_y(p.duration_ms):.1f}"
                for i, p in enumerate(series.points)
            ]
            path_data = " ".join(path_coords)
            paths_svg.append(
                f'<path d="{path_data}" fill="none" stroke="{series.color}" stroke-width="2.5" />'
            )

            # Draw data point circles
            for p in series.points:
                paths_svg.append(
                    f'<circle cx="{scale_x(p.n):.1f}" cy="{scale_y(p.duration_ms):.1f}" '
                    f'r="4" fill="{series.color}" />'
                )

            # Legend entry
            leg_y = padding + 20 + idx * 25
            legends_svg.append(f"""
    <rect x="{width - 240}" y="{leg_y - 10}" width="12" height="12" fill="{series.color}" rx="2" />
    <text x="{width - 220}" y="{leg_y}" fill="#e2e8f0" font-size="12">{series.name}</text>""")

        rendered_paths = "\n  ".join(paths_svg)
        rendered_legends = "".join(legends_svg)

        return f"""<svg viewBox="0 0 {width} {height}" width="100%" height="{height}" xmlns="http://www.w3.org/2000/svg" style="background-color: #0f172a; border-radius: 8px; font-family: 'Courier New', monospace;">
  <text x="{width // 2}" y="32" fill="#f8fafc" font-size="16" font-weight="bold" text-anchor="middle">Empirical Execution Scaling: Time vs Input Size (N)</text>
  
  <!-- Axes -->
  <line x1="{padding}" y1="{height - padding}" x2="{width - padding}" y2="{height - padding}" stroke="#475569" stroke-width="1.5" />
  <line x1="{padding}" y1="{padding}" x2="{padding}" y2="{height - padding}" stroke="#475569" stroke-width="1.5" />
  
  <!-- Labels -->
  <text x="{width // 2}" y="{height - 15}" fill="#94a3b8" font-size="12" text-anchor="middle">Input Size (N Items)</text>
  <text x="20" y="{height // 2}" fill="#94a3b8" font-size="12" text-anchor="middle" transform="rotate(-90 20 {height // 2})">Time (ms)</text>

  <!-- Data Paths -->
  {rendered_paths}

  <!-- Legends -->{rendered_legends}
</svg>"""
````

### 7.9 Summary

In this chapter, we explored the foundations of data organisation and
algorithmic complexity:

1.  **Abstract Data Types (ADTs)**: ADTs separate abstract behavioural
    contracts from physical memory implementations, decoupling interface
    from data layout.
2.  **Asymptotic Complexity**: Big-O notation ($O(f(N))$) provides an
    upper bound on algorithmic growth rates, abstracting away
    hardware-specific timing variations.
3.  **Amortised Analysis**: Geometric capacity doubling bounds the
    average cost of dynamic array insertions to strictly $O(1)$
    amortised time.
4.  **Linear Invariants**: Stacks enforce Last-In, First-Out (LIFO)
    access, while Queues enforce First-In, First-Out (FIFO) access.
5.  **Persistent Okasaki Queues**: Pairing a front list with a reversed
    rear list enables persistent, purely functional queues that execute
    both `enqueue` and `dequeue` in amortised $O(1)$ time while
    maintaining referential transparency.
