### 3.1 The Mystery of the Moving Machine

When you read a piece of source code on a screen, it appears static:
lines of text arranged neatly in a file. Yet, when executed, that frozen
text transforms into a dynamic process. Memory locations are allocated,
variables are instantiated, voltages fluctuate across CPU registers, and
control flows through branches and iterations.

How does a programmer reason about this transformation? (See fig. 1 for
an overview of this process.)

Novices often struggle not because they fail to memorise syntax
(keywords like `let`, `const`, or `function` can be learned in an
afternoon), but because they lack a robust **notional machine**—an
accurate mental model of how the underlying runtime tracks state,
evaluates expressions, and moves from one instruction to the next.

In computer science, two primary models of execution have shaped our
understanding:

1.  **The Dictionary (Environment) Notional Machine**: Code executes by
    sequentially mutating a dictionary of variable-to-value bindings.
    Computation is viewed as a sequence of discrete state transitions
    over a global or scoped memory store.
2.  **The Substitutional (Term Rewriting) Execution Model**: Code
    executes by direct mathematical replacement. Variables are immutable
    names for expressions; execution is the process of reducing
    composite expressions to irreducible literal values by
    systematically substituting identifiers with their definitions.

In this chapter, we will deconstruct program flow by formalising both
notional machines. We will explore the boundary between statements and
expressions, map how environments manage scope, and implement two
complete execution engines in Python: a **step-by-step stateful
environment interpreter** and a **pure algebraic substitution engine**.

### 3.2 Notional Machines: Dual Perspectives on Execution

A **notional machine** is an idealised representation of the computer
hardware and runtime environment designed to explain how programming
language constructs behave. It abstracts away microarchitectural noise
(such as CPU branch predictors, cache hierarchies, and TLB lookups)
while preserving semantic precision.

#### The Dictionary (Environment) Model

Rooted in the Von Neumann architecture and imperative programming, the
Dictionary Model views the computer as an active processing unit paired
with a passive, addressable store:

- The runtime maintains an **Environment** (or Frame Dictionary):
  $\sigma: \text{Identifier} \to \text{Value}$.
- A **Program Counter (PC)** points to the currently active statement in
  a linear sequence.
- Executing an assignment statement $x = e$ evaluates $e$ in the context
  of current state $\sigma$, producing a concrete value $v$, and updates
  the dictionary: $\sigma' = \sigma[x \mapsto v]$.
- Flow is sequential: after statement $S_i$ updates $\sigma$, statement
  $S_{i+1}$ executes under $\sigma'$ (see fig. 2).

#### The Substitutional (Term Rewriting) Model

Rooted in Alonzo Church’s $\lambda$-calculus and mathematical logic, the
Substitutional Model dispenses with state, memory addresses, and time.
There is no mutable dictionary. There is only an expression undergoing
successive reduction steps:

- Variables are not memory containers; they are symbolic placeholders
  for immutable expressions.
- Executing a binding $\text{let } x = e_1 \text{ in } e_2$ proceeds by
  first reducing $e_1$ to an irreducible value $v_1$, and then
  **substituting** all free occurrences of the identifier $x$ inside
  $e_2$ with $v_1$:

<span id="eq:e3-1">$$(\text{let } x = v_1 \text{ in } e_2) \implies e_2[x \mapsto v_1]\qquad{(1)}$$</span>

- Computation terminates when the entire program expression has been
  rewritten into a terminal value.

### 3.4 Operational Semantics: Small-Step vs Big-Step

To formally verify whether an execution engine runs correctly, computer
scientists specify the behavior of programming languages using
**Operational Semantics**.

#### 1. Small-Step Semantics (Reduction Relations)

Small-step operational semantics models individual transition steps of
computation, written as:

<span id="eq:e3-2">$$\langle e, \sigma \rangle \longrightarrow \langle e', \sigma' \rangle\qquad{(2)}$$</span>

This reads: *“Expression $e$ in environment $\sigma$ transitions in a
single atomic step to expression $e'$ under updated environment
$\sigma'$.”*

Small-step semantics is ideal for reasoning about execution traces,
interactive debuggers, multi-threaded concurrency, and intermediate
execution states.

#### 2. Big-Step Semantics (Evaluation Relations)

Big-step operational semantics models the complete evaluation of an
expression directly to its final value in a single relation, written as:

<span id="eq:e3-3">$$\langle e, \sigma \rangle \Downarrow v\qquad{(3)}$$</span>

This reads: *“Expression $e$ in environment $\sigma$ evaluates
completely to terminal value $v$.”* Big-step semantics is concise and
mirrors recursive AST interpreters, but it abstracts away the
intermediate flow of control.

### 3.5 Experimental Implementation in Python

The listings below implement both execution engines in Python to observe
how program flow operates across stateful and substitutional paradigms.

- lst. 1 (`chap03/ast_nodes.py`) defines an Abstract Syntax Tree (AST)
  representation of arbitrarily complex expressions, consisting of
  literals, variables, binary operations, lexical let-bindings,
  assignments, sequences, and blocks.
- lst. 2 (`chap03/environment_machine.py`) implements a Dictionary
  (Environment) notional machine which models execution through
  explicit, scoped environment dictionaries. Each step records an
  immutable snapshot of all visible bindings for diagnostics and
  time-travel debugging.
- lst. 3 (`chap03/substitution_machine.py`) utilizes an alternative
  Substitutional (Term-Rewriting) execution model, which executes
  programs by direct syntactic replacement ($e[x \mapsto v]$), tracking
  each small-step reduction until the expression reaches normal form.
- lst. 4 (`chap03/demo.py`) compares these two engines for equivalence.

Listing 1: Abstract Syntax Tree (AST) Definition (ast_nodes.py)

````
"""
Abstract Syntax Tree (AST) definitions for program flow, bindings, and expressions.
"""

from __future__ import annotations
from dataclasses import dataclass
from typing import Literal, Sequence, TypeAlias

# Primitive terminal values supported by our language
PrimitiveValue: TypeAlias = int | float | bool | str

BinaryOperator: TypeAlias = Literal['+', '-', '*', '/', '==', '<', '&&', '||']

@dataclass(frozen=True, slots=True)
class LiteralExpr:
    value: PrimitiveValue
    kind: Literal['literal'] = 'literal'

@dataclass(frozen=True, slots=True)
class VariableExpr:
    name: str
    kind: Literal['variable'] = 'variable'

@dataclass(frozen=True, slots=True)
class BinaryExpr:
    op: BinaryOperator
    left: Expr
    right: Expr
    kind: Literal['binary'] = 'binary'

@dataclass(frozen=True, slots=True)
class LetExpr:
    name: str
    initializer: Expr
    body: Expr
    kind: Literal['let'] = 'let'

@dataclass(frozen=True, slots=True)
class AssignExpr:
    name: str
    value: Expr
    kind: Literal['assign'] = 'assign'

@dataclass(frozen=True, slots=True)
class SequenceExpr:
    expressions: tuple[Expr, ...]
    kind: Literal['sequence'] = 'sequence'

@dataclass(frozen=True, slots=True)
class BlockExpr:
    body: Expr
    kind: Literal['block'] = 'block'

Expr: TypeAlias = (
    LiteralExpr
    | VariableExpr
    | BinaryExpr
    | LetExpr
    | AssignExpr
    | SequenceExpr
    | BlockExpr
)

class Flow:
    """Smart constructors (factories) for Flow AST nodes."""

    @staticmethod
    def lit(value: PrimitiveValue) -> LiteralExpr:
        return LiteralExpr(value)

    @staticmethod
    def var(name: str) -> VariableExpr:
        return VariableExpr(name)

    @staticmethod
    def bin(op: BinaryOperator, left: Expr, right: Expr) -> BinaryExpr:
        return BinaryExpr(op, left, right)

    @staticmethod
    def let(name: str, initializer: Expr, body: Expr) -> LetExpr:
        return LetExpr(name, initializer, body)

    @staticmethod
    def assign(name: str, value: Expr) -> AssignExpr:
        return AssignExpr(name, value)

    @staticmethod
    def seq(*expressions: Expr) -> SequenceExpr:
        return SequenceExpr(tuple(expressions))

    @staticmethod
    def block(body: Expr) -> BlockExpr:
        return BlockExpr(body)
````

Listing 2: Engine 1: The Dictionary Environment Model
(environment_machine.py)

````
"""
Dictionary / Environment Notional Machine.
Executes programs sequentially by updating variable-to-value bindings across lexical frames.
"""

from __future__ import annotations
from dataclasses import dataclass
from typing import Mapping
#from ast_nodes import Expr, PrimitiveValue, LiteralExpr, VariableExpr, BinaryExpr, LetExpr, AssignExpr, SequenceExpr, BlockExpr

class Environment:
    """
    An Environment Frame represents a scope dictionary with an optional parent pointer.
    """

    def __init__(self, parent: Environment | None = None) -> None:
        self._store: dict[str, PrimitiveValue] = {}
        self.parent: Environment | None = parent

    def define(self, name: str, value: PrimitiveValue) -> None:
        """Declares a new variable in the local scope frame."""
        self._store[name] = value

    def lookup(self, name: str) -> PrimitiveValue:
        """Resolves a variable by walking up the lexical environment chain."""
        if name in self._store:
            return self._store[name]
        if self.parent is not None:
            return self.parent.lookup(name)
        raise NameError(f"Runtime Error: Unbound identifier '{name}'.")

    def assign(self, name: str, value: PrimitiveValue) -> None:
        """Mutates an existing variable binding in the closest enclosing scope frame."""
        if name in self._store:
            self._store[name] = value
            return
        if self.parent is not None:
            self.parent.assign(name, value)
            return
        raise NameError(f"Runtime Error: Cannot assign to undeclared identifier '{name}'.")

    def snapshot(self) -> dict[str, PrimitiveValue]:
        """Produces a snapshot representation of all visible bindings."""
        parent_snapshot = self.parent.snapshot() if self.parent is not None else {}
        return {**parent_snapshot, **self._store}

@dataclass(frozen=True, slots=True)
class StateTraceStep:
    step_number: int
    expression_summary: str
    environment_snapshot: dict[str, PrimitiveValue]
    evaluated_value: PrimitiveValue

class DictionaryStepper:
    """
    Evaluates an AST using the Dictionary Environment Model, recording full state transitions.
    """

    def __init__(self) -> None:
        self._step_count: int = 0
        self._trace_log: list[StateTraceStep] = []

    def get_trace(self) -> tuple[StateTraceStep, ...]:
        return tuple(self._trace_log)

    def evaluate(self, expr: Expr, env: Environment | None = None) -> PrimitiveValue:
        if env is None:
            env = Environment()

        match expr:
            case LiteralExpr(value):
                return value

            case VariableExpr(name):
                val = env.lookup(name)
                self._record_step(f"Lookup '{name}'", env, val)
                return val

            case BinaryExpr(op, left, right):
                left_val = self.evaluate(left, env)
                right_val = self.evaluate(right, env)
                result = self._apply_binary_op(op, left_val, right_val)
                self._record_step(f"Eval Binary: ({left_val} {op} {right_val})", env, result)
                return result

            case LetExpr(name, initializer, body):
                init_val = self.evaluate(initializer, env)
                # Create a new child lexical environment frame for the body
                inner_env = Environment(parent=env)
                inner_env.define(name, init_val)
                self._record_step(f"Bind '{name}' = {init_val}", inner_env, init_val)
                return self.evaluate(body, inner_env)

            case AssignExpr(name, value):
                new_val = self.evaluate(value, env)
                env.assign(name, new_val)
                self._record_step(f"Mutate '{name}' = {new_val}", env, new_val)
                return new_val

            case SequenceExpr(expressions):
                last_value: PrimitiveValue = 0
                for sub_expr in expressions:
                    last_value = self.evaluate(sub_expr, env)
                return last_value

            case BlockExpr(body):
                block_env = Environment(parent=env)
                return self.evaluate(body, block_env)

    def _apply_binary_op(
        self,
        op: str,
        left: PrimitiveValue,
        right: PrimitiveValue,
    ) -> PrimitiveValue:
        # In Python, bool is a subclass of int, so we check bool explicitly first
        if isinstance(left, bool) and isinstance(right, bool):
            match op:
                case '&&': return left and right
                case '||': return left or right
                case '==': return left == right

        elif (
            isinstance(left, (int, float))
            and isinstance(right, (int, float))
            and not isinstance(left, bool)
            and not isinstance(right, bool)
        ):
            match op:
                case '+': return left + right
                case '-': return left - right
                case '*': return left * right
                case '/':
                    if right == 0:
                        raise ZeroDivisionError("Runtime Error: Division by zero.")
                    return (
                        left // right
                        if isinstance(left, int) and isinstance(right, int)
                        else left / right
                    )
                case '==': return left == right
                case '<': return left < right

        if op == '==':
            return left == right

        raise TypeError(
            f"Runtime Error: Illegal binary operation ({type(left).__name__} {op} {type(right).__name__})."
        )

    def _record_step(
        self,
        summary: str,
        env: Environment,
        val: PrimitiveValue,
    ) -> None:
        self._step_count += 1
        self._trace_log.append(
            StateTraceStep(
                step_number=self._step_count,
                expression_summary=summary,
                environment_snapshot=env.snapshot(),
                evaluated_value=val,
            )
        )
````

````
"""
Substitutional / Term-Rewriting Execution Model.
Reduces expressions step-by-step through pure algebraic symbol substitution.
"""

from __future__ import annotations
from dataclasses import dataclass
#from ast_nodes import Expr, PrimitiveValue, LiteralExpr, VariableExpr, BinaryExpr, LetExpr, AssignExpr, SequenceExpr, BlockExpr, Flow

def is_value(expr: Expr) -> bool:
    """Checks whether an AST node is an irreducible terminal literal value."""
    return isinstance(expr, LiteralExpr)

def format_ast(expr: Expr) -> str:
    """Pretty-prints an AST expression into a compact string representation."""
    match expr:
        case LiteralExpr(value):
            return f'"{value}"' if isinstance(value, str) else str(value)
        case VariableExpr(name):
            return name
        case BinaryExpr(op, left, right):
            return f"({format_ast(left)} {op} {format_ast(right)})"
        case LetExpr(name, initializer, body):
            return f"(let {name} = {format_ast(initializer)} in {format_ast(body)})"
        case AssignExpr(name, value):
            return f"({name} := {format_ast(value)})"
        case SequenceExpr(expressions):
            return f"({'; '.join(format_ast(e) for e in expressions)})"
        case BlockExpr(body):
            return f"{{ {format_ast(body)} }}"

def substitute(
    expr: Expr,
    target_var: str,
    replacement: PrimitiveValue,
) -> Expr:
    """
    Performs capture-avoiding syntactic substitution: expr[target_var ↦ replacement].
    Replaces every free occurrence of target_var within expr with replacement.
    """
    match expr:
        case LiteralExpr(_):
            return expr

        case VariableExpr(name):
            return Flow.lit(replacement) if name == target_var else expr

        case BinaryExpr(op, left, right):
            return Flow.bin(
                op,
                substitute(left, target_var, replacement),
                substitute(right, target_var, replacement),
            )

        case LetExpr(name, initializer, body):
            new_init = substitute(initializer, target_var, replacement)
            # If the inner let shadows target_var, do not substitute into the body
            if name == target_var:
                return Flow.let(name, new_init, body)
            return Flow.let(
                name,
                new_init,
                substitute(body, target_var, replacement),
            )

        case AssignExpr(name, value):
            return Flow.assign(name, substitute(value, target_var, replacement))

        case SequenceExpr(expressions):
            return Flow.seq(*(substitute(e, target_var, replacement) for e in expressions))

        case BlockExpr(body):
            return Flow.block(substitute(body, target_var, replacement))

@dataclass(frozen=True, slots=True)
class ReductionStep:
    step_number: int
    before: str
    action: str
    after: str

@dataclass(frozen=True, slots=True)
class StepOutcome:
    next_expr: Expr
    action: str

class SubstitutionStepper:
    """
    Evaluates an AST step-by-step using pure Term Rewriting (Small-Step Operational Semantics).
    """

    def reduce_step(self, expr: Expr) -> StepOutcome | None:
        """
        Executes a single atomic reduction step: e → e'.
        Returns None if the expression is already in normal form (terminal value).
        """
        if is_value(expr):
            return None

        match expr:
            case VariableExpr(name):
                raise NameError(
                    f"Substitutional Model Error: Free variable '{name}' encountered during reduction."
                )

            case BinaryExpr(op, left, right):
                # Left operand must be reduced to a value first
                if not is_value(left):
                    step = self.reduce_step(left)
                    if step is None:
                        raise RuntimeError("Failed to reduce left operand.")
                    return StepOutcome(
                        next_expr=Flow.bin(op, step.next_expr, right),
                        action=f"Reduce Left Operand: {step.action}",
                    )

                # Right operand reduced next
                if not is_value(right):
                    step = self.reduce_step(right)
                    if step is None:
                        raise RuntimeError("Failed to reduce right operand.")
                    return StepOutcome(
                        next_expr=Flow.bin(op, left, step.next_expr),
                        action=f"Reduce Right Operand: {step.action}",
                    )

                # Both sides are values: perform primitive reduction
                assert isinstance(left, LiteralExpr)
                assert isinstance(right, LiteralExpr)
                res = self._apply_primitive_op(op, left.value, right.value)
                return StepOutcome(
                    next_expr=Flow.lit(res),
                    action=f"Compute Primitive ({left.value} {op} {right.value}) → {res}",
                )

            case LetExpr(name, initializer, body):
                # Initializer must reduce to a value first
                if not is_value(initializer):
                    step = self.reduce_step(initializer)
                    if step is None:
                        raise RuntimeError("Failed to reduce let initializer.")
                    return StepOutcome(
                        next_expr=Flow.let(name, step.next_expr, body),
                        action=f"Reduce Initializer for '{name}': {step.action}",
                    )

                # Substitute the evaluated literal into the body
                assert isinstance(initializer, LiteralExpr)
                init_val = initializer.value
                substituted_body = substitute(body, name, init_val)
                return StepOutcome(
                    next_expr=substituted_body,
                    action=f"Substitute [{name} ↦ {init_val}] into body",
                )

            case BlockExpr(body):
                if is_value(body):
                    return StepOutcome(
                        next_expr=body,
                        action=f"Unwrap Block {{ {format_ast(body)} }} → {format_ast(body)}",
                    )
                step = self.reduce_step(body)
                if step is None:
                    raise RuntimeError("Failed to reduce block body.")
                return StepOutcome(
                    next_expr=Flow.block(step.next_expr),
                    action=f"Step inside Block: {step.action}",
                )

            case SequenceExpr(expressions):
                if len(expressions) == 0:
                    return StepOutcome(next_expr=Flow.lit(0), action="Empty sequence evaluates to 0")
                if len(expressions) == 1:
                    return StepOutcome(
                        next_expr=expressions[0],
                        action="Unwrap singleton sequence",
                    )
                first, *rest = expressions
                if is_value(first):
                    return StepOutcome(
                        next_expr=Flow.seq(*rest),
                        action=f"Discard evaluated value {format_ast(first)} in sequence",
                    )
                step = self.reduce_step(first)
                if step is None:
                    raise RuntimeError("Failed to reduce head of sequence.")
                return StepOutcome(
                    next_expr=Flow.seq(step.next_expr, *rest),
                    action=f"Step Head of Sequence: {step.action}",
                )

            case AssignExpr(_, _):
                raise NotImplementedError(
                    "Pure Substitutional Model does not support mutable assignments without state threading."
                )

    def trace_reduction(self, initial_expr: Expr) -> tuple[ReductionStep, ...]:
        """Reduces an expression to normal form, recording the step-by-step reduction trace."""
        steps: list[ReductionStep] = []
        current = initial_expr
        count = 0

        while not is_value(current):
            count += 1
            step = self.reduce_step(current)
            if step is None:
                break

            steps.append(
                ReductionStep(
                    step_number=count,
                    before=format_ast(current),
                    action=step.action,
                    after=format_ast(step.next_expr),
                )
            )
            current = step.next_expr

        return tuple(steps)

    def _apply_primitive_op(
        self,
        op: str,
        left: PrimitiveValue,
        right: PrimitiveValue,
    ) -> PrimitiveValue:
        if isinstance(left, bool) and isinstance(right, bool):
            match op:
                case '&&': return left and right
                case '||': return left or right
                case '==': return left == right

        elif (
            isinstance(left, (int, float))
            and isinstance(right, (int, float))
            and not isinstance(left, bool)
            and not isinstance(right, bool)
        ):
            match op:
                case '+': return left + right
                case '-': return left - right
                case '*': return left * right
                case '/':
                    if right == 0:
                        raise ZeroDivisionError("Division by zero in primitive evaluation.")
                    return (
                        left // right
                        if isinstance(left, int) and isinstance(right, int)
                        else left / right
                    )
                case '==': return left == right
                case '<': return left < right

        raise TypeError(f"Type error in primitive evaluation: {left} {op} {right}")
````

````
"""
Execution harness comparing the Dictionary Notional Machine against pure Term Rewriting.
"""

from __future__ import annotations
import json
#from ast_nodes import Flow
#from environment_machine import DictionaryStepper
#from substitution_machine import SubstitutionStepper, format_ast

def banner(title: str) -> None:
    print('\n' + '=' * 76)
    print(f"  {title.upper()}")
    print('=' * 76)

def run_demonstration() -> None:
    # -------------------------------------------------------------
    # Test Program: Nested Lexical Scoping and Arithmetic
    # let x = 5 in
    #   let y = x * 2 in
    #     (x + y) * 3
    # -------------------------------------------------------------
    program = Flow.let(
        'x',
        Flow.lit(5),
        Flow.let(
            'y',
            Flow.bin('*', Flow.var('x'), Flow.lit(2)),
            Flow.bin(
                '*',
                Flow.bin('+', Flow.var('x'), Flow.var('y')),
                Flow.lit(3),
            ),
        ),
    )

    banner('1. Initial Program AST Structure')
    print(f"Expression: {format_ast(program)}")

    # -------------------------------------------------------------
    # Execution Model A: Dictionary Environment Model
    # -------------------------------------------------------------
    banner('2. Execution via Dictionary Notional Machine')
    dict_stepper = DictionaryStepper()
    dict_result = dict_stepper.evaluate(program)
    dict_trace = dict_stepper.get_trace()

    print(f"Final Evaluated Value: {dict_result}\n")
    print("State Transition Log:")
    for step in dict_trace:
        summary_str = step.expression_summary.ljust(28)
        snapshot_str = json.dumps(step.environment_snapshot)
        print(
            f"  [Step {step.step_number}] Action: {summary_str} | Env Snapshot: {snapshot_str}"
        )

    # -------------------------------------------------------------
    # Execution Model B: Pure Substitutional Term Rewriting
    # -------------------------------------------------------------
    banner('3. Execution via Substitutional Reduction Model')
    subst_stepper = SubstitutionStepper()
    subst_steps = subst_stepper.trace_reduction(program)

    print("Term Rewriting Trace:")
    for step in subst_steps:
        print(f"\n  Step {step.step_number}:")
        print(f"    Before: {step.before}")
        print(f"    Action: {step.action}")
        print(f"    After:  {step.after}")

    banner('4. Cross-Model Semantic Verification')
    final_subst_value = subst_steps[-1].after if subst_steps else None
    print(f"Dictionary Machine Result:     {dict_result}")
    print(f"Substitutional Machine Result: {final_subst_value}")
    print(f"Equivalence Confirmed:         {str(dict_result) == final_subst_value}")

if __name__ == '__main__':
    run_demonstration()
````

### 3.6 Summary

In this chapter, we investigated the mechanics of program flow and
execution models:

1.  **Notional Machines**: An accurate notional machine is required to
    correctly predict how code executes. The Dictionary Model and the
    Substitutional Model offer two distinct paradigms for reasoning
    about flow.
2.  **Dictionary Model (State-Driven)**: Execution advances through
    sequential instructions mutating an environment dictionary
    ($\sigma: \text{Identifier} \to \text{Value}$). Scope is structured
    via chains of parent-pointer environment frames.
3.  **Substitutional Model (Term Rewriting)**: Computation is pure
    algebraic graph reduction. Free variables in expressions are
    systematically substituted with evaluated literals
    ($e[x \mapsto v]$) until a terminal value is reached.
4.  **Equational Reasoning**: The Substitutional model provides strong
    mathematical guarantees: any expression can be replaced with its
    value without altering the program’s observable outcome, a property
    known as **referential transparency**.
