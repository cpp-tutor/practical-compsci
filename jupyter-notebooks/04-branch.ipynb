### 4.1 The Fork in the Road

In Chapter 3, we explored how computation flows down sequential paths,
comparing the mutable state transitions of the Dictionary Notional
Machine with the algebraic reductions of the Substitutional Model. Yet
linear programs are fundamentally limited. A program that executes the
exact same sequence of instructions regardless of its input cannot
adapt, decide, or solve non-trivial problems.

Computation gains its expressive power the moment it can
**bifurcate**—choosing between alternate execution trajectories based on
dynamic truth values (see fig. 1).

Historically, imperative languages inherited branching from
machine-level jump instructions (`JMP`, `JZ`, `JNZ`). This produced the
statement-oriented `if-else` construct: a block that splits execution
flow to execute side effects.

In contrast, modern software engineering and functional programming
model branching as **conditional expressions**: self-contained
operations that evaluate a predicate and yield a concrete *value*.

In this chapter, we will examine the transition from imperative control
statements to pure conditional expressions. We will formalise the
algebraic reduction rules governing conditionals—specifically
**if-True**, **if-False**, and **if-irrelevant**—and construct an AST
transpiler and symbolic branch optimiser in Python capable of
eliminating dead paths and collapsing redundant decision trees.

### 4.2 Statements versus Expressions: The Bifurcation Paradigm

To understand conditional logic, we must distinguish between two
fundamental categories of language grammar: **statements** and
**expressions** (see tbl. 1).

| Property | Imperative Branch Statement (`if / else`) | Functional Conditional Expression (`cond ? X : Y`) |
|:-----------------------|:-----------------------|:-----------------------|
| **Semantic Nature** | Unit of *action* / control-flow redirection | Unit of *computation* / value evaluation |
| **Result** | Produces `void` (no value); relies on side effects | Produces a concrete and d value |
| **Variable Mutation** | Requires pre-declaring mutable variables outside | Direct immutable binding (`const res = cond ? ...`) |
| **Completeness** | `else` branch can be omitted (implicit no-op) | `else` branch is mandatory (must yield a value) |
| **Composability** | Cannot be embedded directly into expressions | Nests arbitrarily inside arithmetic/function calls |

Table 1: Statements Compared to Expressions
{#tbl:statement-vs-expression}

#### The Imperative Anti-Pattern: Mutation via Branching

In imperative programming, branching is frequently used to mutate an
outer variable:

``` python
# Imperative Statement Approach: Mutable, leaked scope, non-atomic

if isPremiumMember and orderTotal > 100:
  discount = 20
else:
  discount = 5
```

This pattern has three design weaknesses:

1.  `discount` must be declared with `let`, leaving it open to
    unintended subsequent mutations.
2.  If the `else` branch is accidentally omitted or contains incomplete
    branches, `discount` remains `undefined` at runtime.
3.  The initialization spans multiple statements, preventing the
    compiler from enforcing immutability.

#### The Functional Paradigm: Pure Expression Branching

Under the expression paradigm, the conditional is an atomic
value-yielding operation:

``` python
# Pure Expression Approach: Immutable, atomic, referentially transparent

from typing import Final

discount: Final = 20 if isPremiumMember and orderTotal > 100 else 5
```

Here, the conditional expression is **referentially transparent**: the
entire ternary construct can be treated as a single sub-expression that
reduces directly to $20$ or $5$.

### 4.3 The Algebraic Rules of Conditional Evaluation

Conditional expressions obey exact equational reduction rules. Let
$\text{if } C \text{ then } E_t \text{ else } E_f$ represent a
conditional expression, where $C$ is the Boolean guard, $E_t$ is the
*consequent* (then-branch), and $E_f$ is the *alternative*
(else-branch).

#### Rule 1: The If-True Reduction

If the guard expression $C$ reduces to the literal constant
$\mathbf{true}$, the alternative branch $E_f$ is discarded entirely, and
the expression reduces to $E_t$:

<span id="eq:e4-1">$$\operatorname{cond}(\mathbf{true}, E_t, E_f) \longrightarrow E_t\qquad{(1)}$$</span>

#### Rule 2: The If-False Reduction

If the guard expression $C$ reduces to the literal constant
$\mathbf{false}$, the consequent branch $E_t$ is discarded, and the
expression reduces to $E_f$:

<span id="eq:e4-2">$$\operatorname{cond}(\mathbf{false}, E_t, E_f) \longrightarrow E_f\qquad{(2)}$$</span>

#### Rule 3: The If-Irrelevant Reduction (Branch Collapse)

If both the consequent and alternative branches evaluate to the exact
same expression $E$, the guard condition $C$ has no bearing on the
outcome. The condition can be eliminated (provided $C$ is pure and free
of side effects):

<span id="eq:e4-3">$$\operatorname{cond}(C, E, E) \longrightarrow E\qquad{(3)}$$</span>

This rule is useful for optimizing compilers. If static analysis
determines that two branches compute identical values, the conditional
jump instruction can be eliminated entirely.

#### Rule 4: The If-Complement Rule

Negating the condition swaps the positions of the consequent and
alternative branches:

<span id="eq:e4-4">$$\operatorname{cond}(\neg C, E_t, E_f) \equiv \operatorname{cond}(C, E_f, E_t)\qquad{(4)}$$</span>

#### Rule 5: Contextual Condition Subsumption

Within the evaluation context of $E_t$, the condition $C$ is known to be
$\mathbf{true}$. Within the evaluation context of $E_f$, the condition
$C$ is known to be $\mathbf{false}$.

Therefore, any nested conditional testing the same guard $C$ can be
simplified:

<span id="eq:e4-5">$$\operatorname{cond}(C, \, \operatorname{cond}(C, A, B), \, D) \longrightarrow \operatorname{cond}(C, A, D)\qquad{(5)}$$</span>

<span id="eq:e4-6">$$\operatorname{cond}(C, \, A, \, \operatorname{cond}(C, B, D)) \longrightarrow \operatorname{cond}(C, A, D)\qquad{(6)}$$</span>

### 4.5 Guarded Execution and Short-Circuit Semantics

A critical property of conditional expressions is that branch evaluation
is **lazy**. In an expression $\operatorname{cond}(C, E_t, E_f)$:

1.  The guard condition $C$ is evaluated first.
2.  If $C \Downarrow \mathbf{true}$, $E_t$ is evaluated and $E_f$ is
    **never** evaluated.
3.  If $C \Downarrow \mathbf{false}$, $E_f$ is evaluated and $E_t$ is
    **never** evaluated.

This lazy evaluation model protects against runtime faults, such as null
pointer dereferences or division by zero:

``` python
# Safe Guarded Execution:
# If divisor == 0.0, the consequent is not evaluated; division by zero never occurs.

safeRatio = 0.0 if divisor == 0.0 else numerator / divisor;
```

If conditional expressions were evaluated eagerly (with both branches
evaluated prior to checking the condition, as with standard function
arguments), the safe expression above would fail whenever
`divisor == 0.0`.

### 4.6 Experimental Implementation in Python

The listings below implement demonstrations in Python to observe how
conditional expressions can be represented, evaluated and simplified.

- lst. 1 (`chap04/ast_nodes.py`) defines an AST with conditional
  expressions. Because `if` is a reserved keyword in Python, the smart
  constructor for conditional expressions is named `if_` (and `stmt_if`
  for imperative statements).
- lst. 2 (`chap04/desugar.py`) provides desugaring of (rewriting)
  imperative statements to functional expressions.
- lst. 3 (`chap04/evaluator.py`) enforces strict guard evaluation and
  short-circuit laziness: only the chosen branch is evaluated, while the
  inactive branch is discarded without execution.
- lst. 4 (`chap04/optimizer.py`) builds an algebraic branch simplifier
  implementing `if-true`, `if-false`, `if-irrelevant`, and nested
  condition propagation:
  - **Rule 1 (If-True)**:
    $\operatorname{cond}(\mathbf{true}, E_t, E_f) \to E_t$
  - **Rule 2 (If-False)**:
    $\operatorname{cond}(\mathbf{false}, E_t, E_f) \to E_f$
  - **Rule 3 (If-Irrelevant)**: $\operatorname{cond}(C, E, E) \to E$
  - **Contextual Subsumption**: Tracking assumed conditions inside
    nested subtrees.
- lst. 5 (`chap04/demo.py`) tests all of the above modules and outputs
  detailed descriptions.

Listing 1: AST Architecture for Conditionals (ast_nodes.py)

````
"""
Abstract Syntax Tree (AST) definitions for branching, conditional expressions,
and imperative statement blocks.
"""

from __future__ import annotations
from dataclasses import dataclass
from typing import Literal, TypeAlias

PrimitiveValue: TypeAlias = int | float | bool | str

BinaryOp: TypeAlias = Literal['+', '-', '*', '/', '==', '!=', '<', '<=', '>', '>=', '&&', '||']

@dataclass(frozen=True, slots=True)
class LiteralExpr:
    value: PrimitiveValue
    kind: Literal['literal'] = 'literal'

@dataclass(frozen=True, slots=True)
class VariableExpr:
    name: str
    kind: Literal['variable'] = 'variable'

@dataclass(frozen=True, slots=True)
class BinaryExpr:
    op: BinaryOp
    left: Expr
    right: Expr
    kind: Literal['binary'] = 'binary'

@dataclass(frozen=True, slots=True)
class LetExpr:
    name: str
    initializer: Expr
    body: Expr
    kind: Literal['let'] = 'let'

@dataclass(frozen=True, slots=True)
class IfExpr:
    """
    Functional Conditional Expression: condition ? then_branch : else_branch
    """
    condition: Expr
    then_branch: Expr
    else_branch: Expr
    kind: Literal['if'] = 'if'

@dataclass(frozen=True, slots=True)
class StatementIfNode:
    """
    Imperative Statement-based If Block (for transpilation/desugaring experiments).
    """
    target_variable: str
    condition: Expr
    then_assign: Expr
    else_assign: Expr
    kind: Literal['statement-if'] = 'statement-if'

Expr: TypeAlias = (
    LiteralExpr
    | VariableExpr
    | BinaryExpr
    | LetExpr
    | IfExpr
    | StatementIfNode
)

class Branch:
    """Smart constructors (factories) for Branch AST nodes."""

    @staticmethod
    def lit(value: PrimitiveValue) -> LiteralExpr:
        return LiteralExpr(value)

    @staticmethod
    def var(name: str) -> VariableExpr:
        return VariableExpr(name)

    @staticmethod
    def bin(op: BinaryOp, left: Expr, right: Expr) -> BinaryExpr:
        return BinaryExpr(op, left, right)

    @staticmethod
    def let(name: str, initializer: Expr, body: Expr) -> LetExpr:
        return LetExpr(name, initializer, body)

    @staticmethod
    def if_(condition: Expr, then_branch: Expr, else_branch: Expr) -> IfExpr:
        return IfExpr(condition, then_branch, else_branch)

    @staticmethod
    def stmt_if(
        target_variable: str,
        condition: Expr,
        then_assign: Expr,
        else_assign: Expr,
    ) -> StatementIfNode:
        return StatementIfNode(target_variable, condition, then_assign, else_assign)
````

Listing 2: Desugaring Imperative Statements to Functional Expressions
(desugar.py)

````
"""
Transpiler lowering imperative statement-level branch blocks into pure
functional expressions.
"""

from __future__ import annotations
#from ast_nodes import Expr, StatementIfNode, Branch

def desugar_statement_if(stmt: StatementIfNode, continuation_body: Expr) -> Expr:
    """
    Transforms imperative conditional statements into equivalent functional let-bindings
    wrapping pure conditional expressions.

    Example transformation:
        let x;
        if (c) { x = A; } else { x = B; }
        rest...

    Becomes:
        let x = (c ? A : B) in rest...
    """
    conditional_expression = Branch.if_(
        stmt.condition,
        stmt.then_assign,
        stmt.else_assign,
    )

    return Branch.let(
        stmt.target_variable,
        conditional_expression,
        continuation_body,
    )
````

Listing 3: Guarded Evaluator with Branch Execution Tracking
(evaluator.py)

````
"""
Guarded conditional expression evaluator with short-circuit verification
and branch telemetry logging.
"""

from __future__ import annotations
from dataclasses import dataclass
from typing import Literal, Mapping, TypeAlias
#from ast_nodes import Expr, PrimitiveValue, LiteralExpr, VariableExpr, BinaryExpr, LetExpr, IfExpr, StatementIfNode

Environment: TypeAlias = Mapping[str, PrimitiveValue]

@dataclass(frozen=True, slots=True)
class BranchEvaluationLog:
    branch_taken: Literal['then', 'else']
    discarded_branch: Expr

class GuardedEvaluator:
    """
    Evaluates expressions with lazy branch guards, ensuring discarded paths
    never trigger runtime faults (e.g. division by zero).
    """

    def __init__(self) -> None:
        self._branch_log: list[BranchEvaluationLog] = []

    def get_branch_history(self) -> tuple[BranchEvaluationLog, ...]:
        return tuple(self._branch_log)

    def evaluate(self, expr: Expr, env: Environment | None = None) -> PrimitiveValue:
        if env is None:
            env = {}

        match expr:
            case LiteralExpr(value):
                return value

            case VariableExpr(name):
                if name not in env:
                    raise NameError(f"Unbound variable '{name}'.")
                return env[name]

            case BinaryExpr(op, left, right):
                left_val = self.evaluate(left, env)
                right_val = self.evaluate(right, env)
                return self._eval_binary(op, left_val, right_val)

            case LetExpr(name, initializer, body):
                init_val = self.evaluate(initializer, env)
                new_env = {**env, name: init_val}
                return self.evaluate(body, new_env)

            case IfExpr(condition, then_branch, else_branch):
                # Evaluate the guard condition strictly
                cond_val = self.evaluate(condition, env)
                if not isinstance(cond_val, bool):
                    raise TypeError(
                        f"Type Error: Conditional guard must evaluate to Boolean, "
                        f"received '{type(cond_val).__name__}' ({cond_val})."
                    )

                if cond_val is True:
                    self._branch_log.append(
                        BranchEvaluationLog(
                            branch_taken='then',
                            discarded_branch=else_branch,
                        )
                    )
                    # Lazy evaluation: only evaluate then_branch
                    return self.evaluate(then_branch, env)
                else:
                    self._branch_log.append(
                        BranchEvaluationLog(
                            branch_taken='else',
                            discarded_branch=then_branch,
                        )
                    )
                    # Lazy evaluation: only evaluate else_branch
                    return self.evaluate(else_branch, env)

            case StatementIfNode():
                raise RuntimeError(
                    "StatementIfNode must be desugared into functional IfExpr prior to evaluation."
                )

    def _eval_binary(
        self,
        op: str,
        left: PrimitiveValue,
        right: PrimitiveValue,
    ) -> PrimitiveValue:
        # In Python, bool is a subclass of int; check bools first
        if isinstance(left, bool) and isinstance(right, bool):
            match op:
                case '&&': return left and right
                case '||': return left or right
                case '==': return left == right
                case '!=': return left != right

        elif (
            isinstance(left, (int, float))
            and isinstance(right, (int, float))
            and not isinstance(left, bool)
            and not isinstance(right, bool)
        ):
            match op:
                case '+': return left + right
                case '-': return left - right
                case '*': return left * right
                case '/':
                    if right == 0:
                        raise ZeroDivisionError("Division by zero.")
                    return (
                        left // right
                        if isinstance(left, int) and isinstance(right, int)
                        else left / right
                    )
                case '==': return left == right
                case '!=': return left != right
                case '<': return left < right
                case '<=': return left <= right
                case '>': return left > right
                case '>=': return left >= right

        if op == '==':
            return left == right
        if op == '!=':
            return left != right

        raise TypeError(
            f"Incompatible types for operator '{op}': "
            f"({type(left).__name__}, {type(right).__name__})."
        )
````

Listing 4: Symbolic Branch Optimiser (Algebraic Tree Rewriter)
(optimizer.py)

````
"""
Symbolic AST branch optimizer applying If-True, If-False, If-Irrelevant,
and contextual condition subsumption reduction rules.
"""

from __future__ import annotations
from dataclasses import dataclass
#from ast_nodes import Expr, LiteralExpr, VariableExpr, BinaryExpr, LetExpr, IfExpr, StatementIfNode, Branch

def are_asts_equal(a: Expr, b: Expr) -> bool:
    """Checks structural equality between two AST expressions."""
    return a == b

@dataclass(frozen=True, slots=True)
class OptimizationMetrics:
    nodes_eliminated: int
    rules_applied: tuple[str, ...]

class BranchOptimizer:
    """
    Optimises an AST by recursively collapsing redundant branches and
    applying algebraic rules until a fixed point is reached.
    """

    def __init__(self) -> None:
        self._eliminated_count: int = 0
        self._applied_rules: list[str] = []

    def get_metrics(self) -> OptimizationMetrics:
        return OptimizationMetrics(
            nodes_eliminated=self._eliminated_count,
            rules_applied=tuple(self._applied_rules),
        )

    def optimize_to_fixed_point(self, expr: Expr) -> Expr:
        """Runs the optimiser to a fixed point (until no further reductions occur)."""
        current = expr
        iterations = 0
        max_iterations = 100

        while iterations < max_iterations:
            initial_count = self._eliminated_count
            current = self._optimize_node(current, frozenset())
            if self._eliminated_count == initial_count:
                break  # Fixed point reached
            iterations += 1

        return current

    def _optimize_node(
        self,
        expr: Expr,
        assumed_true_conditions: frozenset[Expr],
    ) -> Expr:
        match expr:
            case LiteralExpr(_) | VariableExpr(_):
                return expr

            case BinaryExpr(op, left, right):
                return Branch.bin(
                    op,
                    self._optimize_node(left, assumed_true_conditions),
                    self._optimize_node(right, assumed_true_conditions),
                )

            case LetExpr(name, initializer, body):
                return Branch.let(
                    name,
                    self._optimize_node(initializer, assumed_true_conditions),
                    self._optimize_node(body, assumed_true_conditions),
                )

            case IfExpr(condition, then_branch, else_branch):
                # Step 1: Recursively optimize the condition
                opt_cond = self._optimize_node(condition, assumed_true_conditions)

                # Check if condition is already known from enclosing context
                if opt_cond in assumed_true_conditions:
                    self._record_rule("Contextual Subsumption (Known True)", 2)
                    return self._optimize_node(then_branch, assumed_true_conditions)

                # Rule 1: If-True
                if isinstance(opt_cond, LiteralExpr) and opt_cond.value is True:
                    self._record_rule("Rule 1: If-True (Pruned Else-Branch)", 2)
                    return self._optimize_node(then_branch, assumed_true_conditions)

                # Rule 2: If-False
                if isinstance(opt_cond, LiteralExpr) and opt_cond.value is False:
                    self._record_rule("Rule 2: If-False (Pruned Then-Branch)", 2)
                    return self._optimize_node(else_branch, assumed_true_conditions)

                # Context extension for then-branch: condition is known True
                then_assumptions = assumed_true_conditions | {opt_cond}

                opt_then = self._optimize_node(then_branch, then_assumptions)
                opt_else = self._optimize_node(else_branch, assumed_true_conditions)

                # Rule 3: If-Irrelevant (Both branches compute identical subtrees)
                if are_asts_equal(opt_then, opt_else):
                    self._record_rule("Rule 3: If-Irrelevant (Identical Branches Collapsed)", 2)
                    return opt_then

                return Branch.if_(opt_cond, opt_then, opt_else)

            case StatementIfNode():
                return expr

    def _record_rule(self, name: str, nodes_removed: int) -> None:
        self._applied_rules.append(name)
        self._eliminated_count += nodes_removed
````

### 4.7 Summary

In this chapter, we explored the mechanisms of conditional branching:

1.  **Expressions over Statements**: Conditional expressions
    (`cond ? X : Y`) produce concrete values, enabling referential
    transparency and immutability.
2.  **Guarded Lazy Semantics**: Conditionals evaluate the guard
    condition first and strictly discard the unchosen branch, preventing
    side effects and runtime faults in inactive code.
3.  **Core Algebraic Rules**:
    - **If-True / If-False**:
      $\operatorname{cond}(\mathbf{true}, E_t, E_f) \to E_t$ and
      $\operatorname{cond}(\mathbf{false}, E_t, E_f) \to E_f$.
    - **If-Irrelevant**: $\operatorname{cond}(C, E, E) \to E$ eliminates
      redundant branch tests.
    - **Contextual Subsumption**: Conditions known to be true or false
      within nested branches can be simplified.
4.  **AST Optimization**: By running symbolic algebraic reduction passes
    to a fixed point, compilers eliminate dead paths and collapse
    decision trees before runtime.
